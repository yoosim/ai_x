```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.action_chains import ActionChains
from selenium.common.exceptions import TimeoutException, NoSuchElementException
import time
import json
import os
import urllib.request
import re

# === 설정 ===
SCROLL_PAUSE_TIME = 2
BASE_PATH = r"C:\ai_x\source\Project\0709_데이터 폴더"

# 함수 설정
def safe_filename(name):
    return re.sub(r'[\\/*?:"<>|]', "", name)

def download_image(img_url, save_path):
    try:
        headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36',
            'Referer': 'https://map.naver.com/'
        }
        urllib.request.urlopen(urllib.request.Request(img_url, headers=headers))
        urllib.request.urlretrieve(img_url, save_path)
        return save_path
    except Exception as e:
        print(f"이미지 다운로드 실패: {e}")
        return None

# 10개 가게에 대한 브랜드명 리스트 
search_queries = [
    ("쉐이크쉑","쉐이크쉑 신림점"),
    ("맥도날드","맥도날드 신림점"),
    ("롯데리아","롯데리아 신림역점"),
    ("버거킹","버거킹 신림역점"),
    ("버거운버거","버거운버거 서울대점"),
    ("KFC","KFC 신림역"),
    ("노브랜드버거","노브랜드버거 신림남부점"),
    ("프랭크버거","프랭크버거 신림녹두거리점"),
    ("움버거앤윙스","움버거앤윙스 신림역점"),
    ("버거리","버거리 낙성대점")
]   

# 크롬 드라이버 설정
options = webdriver.ChromeOptions()
options.add_argument('user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36')
options.add_argument('window-size=1380,900')
options.add_argument('--no-sandbox')
options.add_argument('--disable-dev-shm-usage')
options.add_argument('--disable-blink-features=AutomationControlled')
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)

driver = webdriver.Chrome(options=options)
driver.execute_script("Object.defineProperty(navigator, 'webdriver', {get: () => undefined})")
wait = WebDriverWait(driver, 15)

# 전체 결과 저장용 딕셔너리
all_menu_data = {}

def click_restaurant_from_list(restaurant_element, index):
    """가게 클릭하여 세부 정보 패널 열기"""
    try:
        # 가게 링크 찾기 - 여러 셀렉터 시도
        link_selectors = [
            "a.place_bluelink",
            "span.place_bluelink", 
            ".place_bluelink",
            "a[class*='place_bluelink']",
            "span[class*='place_bluelink']"
        ]
        
        link_element = None
        for selector in link_selectors:
            try:
                link_element = restaurant_element.find_element(By.CSS_SELECTOR, selector)
                if link_element:
                    break
            except:
                continue
        
        if not link_element:
            print(f"가게 {index} 링크를 찾을 수 없습니다.")
            return False
        
        # 스크롤하여 요소가 보이도록 함
        driver.execute_script("arguments[0].scrollIntoView(true);", link_element)
        time.sleep(1)
        
        # 클릭
        ActionChains(driver).move_to_element(link_element).click().perform()
        print(f"가게 {index} 클릭 완료")
        
        # 세부 정보 패널 로딩 대기
        time.sleep(5)
        return True
        
    except Exception as e:
        print(f"가게 {index} 클릭 실패: {e}")
        return False

def switch_to_detail_iframe():
    """오른쪽 세부 정보 iframe으로 전환"""
    try:
        # 메인으로 돌아가기
        driver.switch_to.default_content()
        time.sleep(2)
        
        # 세부 정보 iframe 찾기
        detail_iframe = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "iframe#entryIframe")))
        driver.switch_to.frame(detail_iframe)
        print("세부 정보 iframe으로 전환 완료")
        time.sleep(3)
        return True
        
    except Exception as e:
        print(f"세부 정보 iframe 전환 실패: {e}")
        return False

def get_restaurant_basic_info():
    """가게 기본 정보 수집"""
    restaurant_info = {}
    
    try:
        # 가게 이름
        name_selectors = ["span.GHAhO", "h1", ".name", "[class*='name']"]
        for selector in name_selectors:
            try:
                name_element = driver.find_element(By.CSS_SELECTOR, selector)
                restaurant_info['name'] = name_element.text.strip()
                print(f"가게명: {restaurant_info['name']}")
                break
            except:
                continue
        
        if 'name' not in restaurant_info:
            restaurant_info['name'] = "가게명 정보 없음"
        
        # 가게 분류
        try:
            category_element = driver.find_element(By.CSS_SELECTOR, "span.lnJFt")
            restaurant_info['category'] = category_element.text.strip()
            print(f"분류: {restaurant_info['category']}")
        except:
            restaurant_info['category'] = "분류 정보 없음"
        
        # 가게 주소
        try:
            address_element = driver.find_element(By.CSS_SELECTOR, "span.LDgIH")
            restaurant_info['address'] = address_element.text.strip()
            print(f"주소: {restaurant_info['address']}")
        except:
            restaurant_info['address'] = "주소 정보 없음"
            
    except Exception as e:
        print(f"기본 정보 수집 실패: {e}")
        restaurant_info = {'name': "정보 수집 실패", 'category': "", 'address': ""}
        
    return restaurant_info

def find_and_click_menu_tab():
    """메뉴 탭을 정확하게 찾아서 클릭"""
    try:
        print("메뉴 탭 찾는 중...")
        
        # 페이지 스크롤하여 탭이 보이도록 함
        driver.execute_script("window.scrollTo(0, 0);")
        time.sleep(2)
        
        # 1단계: 모든 탭 요소들 찾기
        tab_selectors = [
            "a._tab-menu",
            "div.YYh8o a",
            "div[class*='tab'] a",
            "nav a",
            "ul li a",
            "a[href*='menu']",
            "button[class*='tab']"
        ]
        
        all_tabs = []
        for selector in tab_selectors:
            try:
                tabs = driver.find_elements(By.CSS_SELECTOR, selector)
                if tabs:
                    all_tabs.extend(tabs)
            except:
                continue
        
        if not all_tabs:
            print("탭 요소를 찾을 수 없습니다")
            return False
        
        print(f"총 {len(all_tabs)}개의 탭 요소 발견")
        
        # 2단계: 각 탭의 텍스트와 URL 확인하여 메뉴 탭 찾기
        menu_tab = None
        for i, tab in enumerate(all_tabs):
            try:
                # 탭 텍스트 확인
                tab_text = tab.text.strip().lower()
                # href 속성 확인
                tab_href = tab.get_attribute("href") or ""
                
                print(f"탭 {i+1}: 텍스트='{tab_text}', href='{tab_href[-50:]}'")
                
                # 메뉴 탭인지 판단
                is_menu_tab = (
                    "menu" in tab_href.lower() or
                    "메뉴" in tab_text or
                    "menu" in tab_text
                )
                
                if is_menu_tab:
                    menu_tab = tab
                    print(f"메뉴 탭 발견: {tab_text}")
                    break
                    
            except Exception as e:
                print(f"탭 {i+1} 확인 중 오류: {e}")
                continue
        
        # 3단계: 메뉴 탭 클릭
        if menu_tab:
            try:
                # 스크롤하여 요소가 보이도록 함
                driver.execute_script("arguments[0].scrollIntoView(true);", menu_tab)
                time.sleep(2)
                
                # JavaScript로 클릭
                driver.execute_script("arguments[0].click();", menu_tab)
                time.sleep(5)  # 메뉴 로딩 대기 시간 증가
                print("메뉴 탭 클릭 완료")
                return True
                
            except Exception as e:
                print(f"메뉴 탭 클릭 실패: {e}")
                return False
        else:
            print("메뉴 탭을 찾을 수 없습니다")
            return False
            
    except Exception as e:
        print(f"메뉴 탭 찾기 실패: {e}")
        return False

def get_menu_list(driver, brand_name):
    menu_data = []

    # 페이지 로딩 대기
    time.sleep(5)

    # 더보기 클릭 (선택적으로) - 개선된 버전
    print("메뉴 더보기 버튼 찾는 중...")
    more_selectors = [
        "a.fvwqf", 
        "button.fvwqf", 
        "div.fvwqf",
        "a[class*='more']", 
        "button[class*='more']",
        "//a[contains(text(), '더보기')]",
        "//button[contains(text(), '더보기')]",
        "//div[contains(text(), '더보기')]"
    ]
    
    for selector in more_selectors:
        try:
            if selector.startswith("//"):
                btn = driver.find_element(By.XPATH, selector)
            else:
                btn = driver.find_element(By.CSS_SELECTOR, selector)
                
            if btn.is_displayed() and btn.is_enabled():
                driver.execute_script("arguments[0].scrollIntoView(true);", btn)
                time.sleep(1)
                driver.execute_script("arguments[0].click();", btn)
                time.sleep(3)
                print("[+] 더보기 클릭 완료")
                break
        except:
            continue

    # 메뉴 항목들 수집 - 여러 가지 셀렉터 시도
    menu_elements = []
    menu_selectors = [
        "li.E2jtL",
        ".menu-item",
        "li[class*='menu']",
        "div[class*='menu']",
        "li[class*='item']",
        ".item"
    ]
    
    for selector in menu_selectors:
        try:
            elements = driver.find_elements(By.CSS_SELECTOR, selector)
            if elements:
                menu_elements = elements
                print(f"[+] 메뉴 요소 발견: {selector}, 개수: {len(elements)}")
                break
        except:
            continue
    
    if not menu_elements:
        print(f"[!] {brand_name} 메뉴 요소를 찾을 수 없습니다.")
        # 페이지 소스에서 메뉴 관련 요소 디버깅
        try:
            page_source = driver.page_source
            if "메뉴" in page_source:
                print("[DEBUG] 페이지에 '메뉴' 텍스트가 있지만 요소를 찾을 수 없습니다.")
            else:
                print("[DEBUG] 페이지에 메뉴 관련 내용이 없는 것 같습니다.")
        except:
            pass
        return menu_data
    
    for i, el in enumerate(menu_elements):
        try:
            # 메뉴명 추출 - 여러 셀렉터 시도
            name = None
            name_selectors = ["span.lPzHi", ".menu-name", "span.lPzHi", "[class*='name']", "span", "h3", "h4", "div"]
            for name_sel in name_selectors:
                try:
                    name_el = el.find_element(By.CSS_SELECTOR, name_sel)
                    name = name_el.text.strip()
                    if name and len(name) > 1:
                        break
                except:
                    continue
            
            if not name:
                name = f"메뉴_{i+1}"

            # 가격 추출 - 여러 셀렉터 시도
            price = None
            price_selectors = ["div.GXS1X em", ".price", "[class*='price']", "em", "span[class*='won']", "div.GXS1X"]
            for price_sel in price_selectors:
                try:
                    price_el = el.find_element(By.CSS_SELECTOR, price_sel)
                    price = price_el.text.strip()
                    if price and ("원" in price or "," in price or price.isdigit()):
                        break
                except:
                    continue
            
            if not price:
                price = "가격 정보 없음"

            # 설명 추출 - 여러 셀렉터 시도
            desc = None
            desc_selectors = ["div.TRxGt", ".description", "[class*='desc']", "p"]
            for desc_sel in desc_selectors:
                try:
                    desc_el = el.find_element(By.CSS_SELECTOR, desc_sel)
                    desc = desc_el.text.strip()
                    if desc and len(desc) > 3:
                        break
                except:
                    continue
            
            if not desc:
                desc = "설명 없음"

            # 이미지 URL 추출 - 여러 셀렉터 시도
            img_url = None
            img_selectors = ["img.K0PDV", "img", "[class*='img'] img"]
            for img_sel in img_selectors:
                try:
                    img = el.find_element(By.CSS_SELECTOR, img_sel)
                    img_url = img.get_attribute("src")
                    if img_url and ("http" in img_url):
                        break
                except:
                    continue

            # 이미지 저장 처리
            image_path = None
            if img_url:
                # 메뉴명 기준 하위 폴더 생성
                safe_menu_name = safe_filename(name)[:30]
                menu_folder = os.path.join(BASE_PATH, brand_name, safe_menu_name)
                os.makedirs(menu_folder, exist_ok=True)

                # 파일명 생성 (brand_001.jpg 형식)
                img_filename = f"brand_{(i+1):03d}.jpg"
                save_path = os.path.join(menu_folder, img_filename)

                image_path = download_image(img_url, save_path)

            # 최종 저장
            menu_data.append({
                "menu": name,
                "price": price,
                "description": desc,
                "image_url": img_url,
                "image_path": image_path
            })
            print(f"  ↳ {i+1}. {name} | {price}")

        except Exception as e:
            print(f"[!] 메뉴 항목 {i+1} 처리 중 오류: {e}")
            continue

    return menu_data

def go_back_to_search_list():
    """검색 목록으로 돌아가기"""
    try:
        # 메인으로 돌아가기
        driver.switch_to.default_content()
        time.sleep(2)
        
        # 검색 결과 iframe으로 다시 전환
        search_iframe = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "iframe#searchIframe")))
        driver.switch_to.frame(search_iframe)
        time.sleep(2)
        print("검색 목록으로 돌아가기 완료")
        return True
        
    except Exception as e:
        print(f"검색 목록으로 돌아가기 실패: {e}")
        return False

# 메인 크롤링 실행
for brand_name, query in search_queries:
    print(f"-------[+] {brand_name} 크롤링 시작!-------")
    
    try:
        # 네이버 지도 검색 URL 진입
        search_url = f"https://map.naver.com/p/search/{query}"
        driver.get(search_url)
        time.sleep(5)  # 페이지 로딩 대기 시간 증가
        
        # 검색 결과 iframe 접근
        print(f"-------{query} 검색 결과 iframe으로 전환 중-------")
        try:
            search_iframe = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "iframe#searchIframe")))
            driver.switch_to.frame(search_iframe)
            time.sleep(3)
            
            # 첫 번째 검색 결과 클릭 - 여러 가지 셀렉터 시도
            first_result = None
            selectors = [
                "span.place_bluelink",
                "a.place_bluelink", 
                ".place_bluelink",
                "span[class*='place_bluelink']",
                "a[class*='place_bluelink']",
                "li.UEzoS a",
                "li[class*='item'] a"
            ]
            
            for selector in selectors:
                try:
                    results = driver.find_elements(By.CSS_SELECTOR, selector)
                    if results:
                        first_result = results[0]  # 첫 번째 결과 선택
                        print(f"[+] 검색 결과 찾음: {selector}")
                        break
                except:
                    continue
            
            if not first_result:
                print(f"[!] {brand_name} 검색 결과를 찾을 수 없습니다.")
                all_menu_data[brand_name] = []
                continue
            
            # 검색 결과 클릭
            if not click_restaurant_from_list(first_result, 1):
                all_menu_data[brand_name] = []
                continue
            
            # 상세 정보 iframe으로 전환
            if not switch_to_detail_iframe():
                print(f"[!] {brand_name} 상세 페이지 iframe을 찾을 수 없습니다.")
                all_menu_data[brand_name] = []
                continue
            
            # 기본 정보 수집
            restaurant_info = get_restaurant_basic_info()
            
            # 메뉴 탭 클릭 및 메뉴 정보 수집
            print('-------메뉴 탭 찾는중-------')
            if find_and_click_menu_tab():
                print("메뉴 정보 수집 시작...")
                menu_data = get_menu_list(driver, brand_name)
                all_menu_data[brand_name] = {
                    'restaurant_info': restaurant_info,
                    'menu_data': menu_data
                }
                print(f"[+] {brand_name} 메뉴 {len(menu_data)}개 수집 완료")
            else:
                print(f"[!] {brand_name} 메뉴 탭을 찾을 수 없습니다.")
                all_menu_data[brand_name] = {
                    'restaurant_info': restaurant_info,
                    'menu_data': []
                }
            
        except (TimeoutException, NoSuchElementException) as e:
            print(f"[!] {brand_name} 처리 중 오류 발생: {e}")
            all_menu_data[brand_name] = []
            
    except Exception as e:
        print(f"[!] {brand_name} 크롤링 실패: {e}")
        all_menu_data[brand_name] = []
    
    # 다음 검색을 위해 새 페이지로 이동
    print("다음 검색 준비 중...")
    time.sleep(3)

# 최종 결과 정리
print("\n=== 최종 결과 ===")
for brand_name, data in all_menu_data.items():
    if isinstance(data, dict):
        menu_count = len(data.get('menu_data', []))
        print(f"{brand_name}: {menu_count}개 메뉴")
    else:
        print(f"{brand_name}: 0개 메뉴 (수집 실패)")

# 각 브랜드별로 리스트 분리
for brand_name, data in all_menu_data.items():
    if isinstance(data, dict) and data.get('menu_data'):
        print(f"\n--- {brand_name} 리스트 ---")
        menu_list = [item['menu'] for item in data['menu_data']]
        price_list = [item['price'] for item in data['menu_data']]
        desc_list = [item['description'] for item in data['menu_data']]
        
        print(f"메뉴: {menu_list}")
        print(f"가격: {price_list}")
        print(f"설명: {desc_list}")

# 결과를 JSON 파일로 저장
with open(os.path.join(BASE_PATH, "menu_data.json"), 'w', encoding='utf-8') as f:
    json.dump(all_menu_data, f, ensure_ascii=False, indent=2)

print("\n결과가 menu_data.json 파일로 저장되었습니다.")

# 브라우저 종료
# driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.action_chains import ActionChains
from selenium.common.exceptions import TimeoutException, NoSuchElementException
import time
import json
import os
import urllib.request
import re

# === 설정 ===
SCROLL_PAUSE_TIME = 2
BASE_PATH = r"C:\ai_x\source\Project\0709_데이터 폴더"

# 함수 설정
def safe_filename(name):
    return re.sub(r'[\\/*?:"<>|]', "", name)

def download_image(img_url, save_path):
    try:
        headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36',
            'Referer': 'https://map.naver.com/'
        }
        urllib.request.urlopen(urllib.request.Request(img_url, headers=headers))
        urllib.request.urlretrieve(img_url, save_path)
        return save_path
    except Exception as e:
        print(f"이미지 다운로드 실패: {e}")
        return None

# 10개 가게에 대한 브랜드명 리스트 
search_queries = [
    ("쉐이크쉑","쉐이크쉑 신림점"),
    ("맥도날드","맥도날드 신림점"),
    ("롯데리아","롯데리아 신림역점"),
    ("버거킹","버거킹 신림역점"),
    ("버거운버거","버거운버거 서울대점"),
    ("KFC","KFC 신림역"),
    ("노브랜드버거","노브랜드버거 신림남부점"),
    ("프랭크버거","프랭크버거 신림녹두거리점"),
    ("움버거앤윙스","움버거앤윙스 신림역점"),
    ("버거리","버거리 낙성대점")
]   

# 크롬 드라이버 설정
options = webdriver.ChromeOptions()
options.add_argument('user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36')
options.add_argument('window-size=1380,900')
options.add_argument('--no-sandbox')
options.add_argument('--disable-dev-shm-usage')
options.add_argument('--disable-blink-features=AutomationControlled')
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)

driver = webdriver.Chrome(options=options)
driver.execute_script("Object.defineProperty(navigator, 'webdriver', {get: () => undefined})")
wait = WebDriverWait(driver, 15)

# 전체 결과 저장용 딕셔너리
all_menu_data = {}

def click_restaurant_from_list(restaurant_element, index):
    """가게 클릭하여 세부 정보 패널 열기"""
    try:
        # 가게 링크 찾기 - 여러 셀렉터 시도
        link_selectors = [
            "a.place_bluelink",
            "span.place_bluelink", 
            ".place_bluelink",
            "a[class*='place_bluelink']",
            "span[class*='place_bluelink']"
        ]
        
        link_element = None
        for selector in link_selectors:
            try:
                link_element = restaurant_element.find_element(By.CSS_SELECTOR, selector)
                if link_element:
                    break
            except:
                continue
        
        if not link_element:
            print(f"가게 {index} 링크를 찾을 수 없습니다.")
            return False
        
        # 스크롤하여 요소가 보이도록 함
        driver.execute_script("arguments[0].scrollIntoView(true);", link_element)
        time.sleep(1)
        
        # 클릭
        ActionChains(driver).move_to_element(link_element).click().perform()
        print(f"가게 {index} 클릭 완료")
        
        # 세부 정보 패널 로딩 대기
        time.sleep(5)
        return True
        
    except Exception as e:
        print(f"가게 {index} 클릭 실패: {e}")
        return False

def switch_to_detail_iframe():
    """오른쪽 세부 정보 iframe으로 전환"""
    try:
        # 메인으로 돌아가기
        driver.switch_to.default_content()
        time.sleep(2)
        
        # 세부 정보 iframe 찾기
        detail_iframe = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "iframe#entryIframe")))
        driver.switch_to.frame(detail_iframe)
        print("세부 정보 iframe으로 전환 완료")
        time.sleep(3)
        return True
        
    except Exception as e:
        print(f"세부 정보 iframe 전환 실패: {e}")
        return False

def get_restaurant_basic_info():
    """가게 기본 정보 수집"""
    restaurant_info = {}
    
    try:
        # 가게 이름
        name_selectors = ["span.GHAhO", "h1", ".name", "[class*='name']"]
        for selector in name_selectors:
            try:
                name_element = driver.find_element(By.CSS_SELECTOR, selector)
                restaurant_info['name'] = name_element.text.strip()
                print(f"가게명: {restaurant_info['name']}")
                break
            except:
                continue
        
        if 'name' not in restaurant_info:
            restaurant_info['name'] = "가게명 정보 없음"
        
        # 가게 분류
        try:
            category_element = driver.find_element(By.CSS_SELECTOR, "span.lnJFt")
            restaurant_info['category'] = category_element.text.strip()
            print(f"분류: {restaurant_info['category']}")
        except:
            restaurant_info['category'] = "분류 정보 없음"
        
        # 가게 주소
        try:
            address_element = driver.find_element(By.CSS_SELECTOR, "span.LDgIH")
            restaurant_info['address'] = address_element.text.strip()
            print(f"주소: {restaurant_info['address']}")
        except:
            restaurant_info['address'] = "주소 정보 없음"
            
    except Exception as e:
        print(f"기본 정보 수집 실패: {e}")
        restaurant_info = {'name': "정보 수집 실패", 'category': "", 'address': ""}
        
    return restaurant_info

def find_and_click_menu_tab():
    """메뉴 탭을 정확하게 찾아서 클릭 - 모든 탭 확인"""
    try:
        print("메뉴 탭 찾는 중...")
        
        # 페이지 스크롤하여 탭이 보이도록 함
        driver.execute_script("window.scrollTo(0, 0);")
        time.sleep(1)
        
        # 1단계: 실제 메뉴 탭 셀렉터 사용
        tab_selectors = [
            "a.tpj9w._tab-menu",  # 사용자가 제공한 정확한 클래스
            "a._tab-menu",        # _tab-menu 클래스만으로도 시도
            "a.tpj9w",            # tpj9w 클래스만으로도 시도
            "div.YYh8o a",
            "div[class*='tab'] a",
            "nav a",
            "ul li a"
        ]
        
        all_tabs = []
        for selector in tab_selectors:
            try:
                tabs = driver.find_elements(By.CSS_SELECTOR, selector)
                if tabs:
                    all_tabs = tabs
                    print(f"탭 요소들 찾음: {selector} - {len(tabs)}개")
                    break
            except:
                continue
        
        if not all_tabs:
            print("탭 요소를 찾을 수 없습니다")
            return False
        
        # 2단계: 모든 탭의 정보를 출력하고 메뉴 탭 찾기
        print("=== 모든 탭 정보 확인 ===")
        menu_tab = None
        for i, tab in enumerate(all_tabs):
            try:
                # 탭 텍스트 확인
                tab_text = tab.text.strip()
                # href 속성 확인
                tab_href = tab.get_attribute("href") or ""
                # class 속성 확인
                tab_class = tab.get_attribute("class") or ""
                
                print(f"탭 {i+1}: 텍스트='{tab_text}', class='{tab_class}', href='{tab_href[-50:] if len(tab_href) > 50 else tab_href}'")
                
                # 메뉴 탭인지 판단 - 정확한 조건 적용
                is_menu_tab = (
                    "메뉴" in tab_text or
                    "menu" in tab_text.lower() or
                    "menu" in tab_href.lower()
                )
                
                if is_menu_tab:
                    menu_tab = tab
                    print(f"*** 메뉴 탭 발견: '{tab_text}' (class: {tab_class}) ***")
                    break
                    
            except Exception as e:
                print(f"탭 {i+1} 확인 중 오류: {e}")
                continue
        
        print("=== 탭 정보 확인 완료 ===")
        
        # 3단계: 메뉴 탭 클릭
        if menu_tab:
            try:
                # 스크롤하여 요소가 보이도록 함
                driver.execute_script("arguments[0].scrollIntoView(true);", menu_tab)
                time.sleep(1)
                
                # JavaScript로 클릭
                driver.execute_script("arguments[0].click();", menu_tab)
                time.sleep(5)  # 메뉴 로딩 충분히 기다리기
                print("메뉴 탭 클릭 완료")
                return True
                
            except Exception as e:
                print(f"메뉴 탭 클릭 실패: {e}")
                return False
        else:
            print("메뉴 탭을 찾을 수 없습니다")
            # 메뉴 탭을 찾을 수 없을 때 모든 탭을 한번 더 클릭해보기
            print("모든 탭을 순차적으로 확인해보겠습니다...")
            for i, tab in enumerate(all_tabs):
                try:
                    tab_text = tab.text.strip()
                    print(f"탭 {i+1} '{tab_text}' 클릭해보기...")
                    driver.execute_script("arguments[0].click();", tab)
                    time.sleep(3)
                    
                    # 메뉴 요소가 있는지 확인
                    menu_elements = driver.find_elements(By.CSS_SELECTOR, "li.E2jtL")
                    if menu_elements:
                        print(f"탭 {i+1} '{tab_text}'에서 메뉴 요소 {len(menu_elements)}개 발견!")
                        return True
                except Exception as e:
                    print(f"탭 {i+1} 클릭 실패: {e}")
                    continue
            return False
            
    except Exception as e:
        print(f"메뉴 탭 찾기 실패: {e}")
        return False

def get_menu_list(driver, brand_name):
    """메뉴 탭에서 메뉴 정보 수집 - 새로운 구조에 맞게 수정"""
    menu_data = []

    try:
        print("메뉴 페이지 로딩 대기 중...")
        time.sleep(3)
        
        # 메뉴 더보기 버튼 클릭 (있다면)
        try:
            more_button_selectors = [
                "a.fvwqf",
                "button.fvwqf",
                "a[class*='more']",
                "button[class*='more']",
                "//a[contains(text(), '더보기')]",
                "//button[contains(text(), '더보기')]"
            ]
            
            for selector in more_button_selectors:
                try:
                    if selector.startswith("//"):
                        more_button = driver.find_element(By.XPATH, selector)
                    else:
                        more_button = driver.find_element(By.CSS_SELECTOR, selector)
                    
                    if more_button.is_displayed() and more_button.is_enabled():
                        driver.execute_script("arguments[0].scrollIntoView(true);", more_button)
                        time.sleep(1)
                        driver.execute_script("arguments[0].click();", more_button)
                        time.sleep(2)
                        print("메뉴 더보기 버튼 클릭 완료")
                        break
                except:
                    continue
        except:
            print("메뉴 더보기 버튼이 없거나 이미 모든 메뉴가 표시됨")
        
        # 메뉴 전체 영역 찾기
        try:
            menu_container = driver.find_element(By.CSS_SELECTOR, "ul.order_list_area")
            print("메뉴 컨테이너 찾음")
        except:
            print("메뉴 컨테이너를 찾을 수 없습니다")
            return menu_data
        
        # 메뉴 카테고리별로 처리
        category_sections = driver.find_elements(By.CSS_SELECTOR, "div.order_list_tit")
        print(f"메뉴 카테고리 섹션 {len(category_sections)}개 발견")
        
        for section_idx, section in enumerate(category_sections):
            try:
                # 카테고리 제목 확인
                category_title_element = section.find_element(By.CSS_SELECTOR, "span.title")
                category_title = category_title_element.text.strip()
                print(f"카테고리 {section_idx + 1}: {category_title}")
                
                # 음료, 굿즈 카테고리는 건너뛰기
                if category_title in ["음료", "굿즈"]:
                    print(f"카테고리 '{category_title}' 건너뛰기")
                    continue
                
                # 해당 카테고리 다음에 있는 메뉴 항목들 찾기
                # 카테고리 섹션 다음에 나오는 메뉴 항목들을 찾기 위해 전체 ul에서 검색
                menu_items = menu_container.find_elements(By.CSS_SELECTOR, "li")
                
                # 현재 카테고리 섹션의 위치를 기준으로 메뉴 항목 찾기
                section_found = False
                for item in menu_items:
                    try:
                        # 현재 아이템이 카테고리 섹션인지 확인
                        if not section_found:
                            try:
                                item_category = item.find_element(By.CSS_SELECTOR, "div.order_list_tit span.title")
                                if item_category.text.strip() == category_title:
                                    section_found = True
                                    continue
                            except:
                                pass
                        
                        # 카테고리 섹션을 찾은 후 메뉴 항목들 처리
                        if section_found:
                            # 다음 카테고리 섹션이 나오면 중단
                            try:
                                next_category = item.find_element(By.CSS_SELECTOR, "div.order_list_tit span.title")
                                if next_category.text.strip() != category_title:
                                    break
                            except:
                                pass
                            
                            # 메뉴 항목 정보 추출
                            try:
                                menu_info = {}
                                
                                # 메뉴명 추출
                                try:
                                    name_element = item.find_element(By.CSS_SELECTOR, "div.MenuContent__tit__313LA")
                                    menu_info['menu'] = name_element.text.strip()
                                except:
                                    continue  # 메뉴명이 없으면 메뉴 항목이 아님
                                
                                # 메뉴 설명 추출
                                try:
                                    desc_element = item.find_element(By.CSS_SELECTOR, "div.MenuContent__detail__OG635 span.detail_txt")
                                    menu_info['description'] = desc_element.text.strip()
                                except:
                                    menu_info['description'] = "설명 없음"
                                
                                # 가격 추출
                                try:
                                    price_element = item.find_element(By.CSS_SELECTOR, "div.MenuContent__price__lhCy9")
                                    menu_info['price'] = price_element.text.strip()
                                except:
                                    menu_info['price'] = "가격 정보 없음"
                                
                                # 이미지 추출
                                try:
                                    img_container = item.find_element(By.CSS_SELECTOR, "div.MenuContent__info_img__JnyVE")
                                    img_element = img_container.find_element(By.CSS_SELECTOR, "img[alt='섬네일']")
                                    img_url = img_element.get_attribute("src")
                                    menu_info['image_url'] = img_url
                                    
                                    if img_url:
                                        # 이미지 저장
                                        safe_menu_name = safe_filename(menu_info['menu'])[:30]
                                        menu_folder = os.path.join(BASE_PATH, brand_name, safe_menu_name)
                                        os.makedirs(menu_folder, exist_ok=True)
                                        
                                        img_filename = f"brand_{len(menu_data)+1:03d}.jpg"
                                        save_path = os.path.join(menu_folder, img_filename)
                                        
                                        image_path = download_image(img_url, save_path)
                                        menu_info['image_path'] = image_path
                                    else:
                                        menu_info['image_path'] = None
                                except:
                                    menu_info['image_url'] = None
                                    menu_info['image_path'] = None
                                
                                # 카테고리 정보 추가
                                menu_info['category'] = category_title
                                
                                menu_data.append(menu_info)
                                print(f"  [{category_title}] {len(menu_data)}. {menu_info['menu']} - {menu_info['price']}")
                                
                            except Exception as e:
                                print(f"메뉴 항목 처리 중 오류: {e}")
                                continue
                    except Exception as e:
                        continue
                        
            except Exception as e:
                print(f"카테고리 섹션 {section_idx + 1} 처리 중 오류: {e}")
                continue
        
        # 대체 방법: 직접 메뉴 항목들 찾기 (위 방법이 실패할 경우)
        if len(menu_data) == 0:
            print("대체 방법으로 메뉴 수집 시도...")
            
            # 메뉴명이 있는 모든 항목 찾기
            menu_items = driver.find_elements(By.CSS_SELECTOR, "div.MenuContent__tit__313LA")
            print(f"메뉴 항목 {len(menu_items)}개 발견")
            
            for idx, menu_item in enumerate(menu_items):
                try:
                    # 현재 메뉴 항목의 부모 요소에서 정보 추출
                    parent_element = menu_item.find_element(By.XPATH, "./ancestor::li[1]")
                    
                    # 해당 카테고리가 음료나 굿즈인지 확인
                    try:
                        # 현재 메뉴 항목 위에 있는 카테고리 찾기
                        all_elements = driver.find_elements(By.CSS_SELECTOR, "div.order_list_tit span.title, div.MenuContent__tit__313LA")
                        current_category = None
                        
                        for i, element in enumerate(all_elements):
                            if element == menu_item:
                                # 현재 메뉴 항목보다 앞에 있는 카테고리 찾기
                                for j in range(i-1, -1, -1):
                                    try:
                                        if "order_list_tit" in all_elements[j].find_element(By.XPATH, "./parent::*").get_attribute("class"):
                                            current_category = all_elements[j].text.strip()
                                            break
                                    except:
                                        continue
                                break
                        
                        if current_category in ["음료", "굿즈"]:
                            print(f"메뉴 '{menu_item.text.strip()}' - 카테고리 '{current_category}' 건너뛰기")
                            continue
                            
                    except:
                        current_category = "기타"
                    
                    menu_info = {}
                    
                    # 메뉴명
                    menu_info['menu'] = menu_item.text.strip()
                    
                    # 메뉴 설명
                    try:
                        desc_element = parent_element.find_element(By.CSS_SELECTOR, "div.MenuContent__detail__OG635 span.detail_txt")
                        menu_info['description'] = desc_element.text.strip()
                    except:
                        menu_info['description'] = "설명 없음"
                    
                    # 가격
                    try:
                        price_element = parent_element.find_element(By.CSS_SELECTOR, "div.MenuContent__price__lhCy9")
                        menu_info['price'] = price_element.text.strip()
                    except:
                        menu_info['price'] = "가격 정보 없음"
                    
                    # 이미지
                    try:
                        img_container = parent_element.find_element(By.CSS_SELECTOR, "div.MenuContent__info_img__JnyVE")
                        img_element = img_container.find_element(By.CSS_SELECTOR, "img[alt='섬네일']")
                        img_url = img_element.get_attribute("src")
                        menu_info['image_url'] = img_url
                        
                        if img_url:
                            safe_menu_name = safe_filename(menu_info['menu'])[:30]
                            menu_folder = os.path.join(BASE_PATH, brand_name, safe_menu_name)
                            os.makedirs(menu_folder, exist_ok=True)
                            
                            img_filename = f"brand_{len(menu_data)+1:03d}.jpg"
                            save_path = os.path.join(menu_folder, img_filename)
                            
                            image_path = download_image(img_url, save_path)
                            menu_info['image_path'] = image_path
                        else:
                            menu_info['image_path'] = None
                    except:
                        menu_info['image_url'] = None
                        menu_info['image_path'] = None
                    
                    menu_info['category'] = current_category or "기타"
                    
                    menu_data.append(menu_info)
                    print(f"  [{menu_info['category']}] {len(menu_data)}. {menu_info['menu']} - {menu_info['price']}")
                    
                except Exception as e:
                    print(f"메뉴 항목 {idx + 1} 처리 중 오류: {e}")
                    continue
                
    except Exception as e:
        print(f"메뉴 정보 수집 실패: {e}")
    
    print(f"총 {len(menu_data)}개 메뉴 수집 완료")
    return menu_data

def go_back_to_search_list():
    """검색 목록으로 돌아가기"""
    try:
        # 메인으로 돌아가기
        driver.switch_to.default_content()
        time.sleep(2)
        
        # 검색 결과 iframe으로 다시 전환
        search_iframe = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "iframe#searchIframe")))
        driver.switch_to.frame(search_iframe)
        time.sleep(2)
        print("검색 목록으로 돌아가기 완료")
        return True
        
    except Exception as e:
        print(f"검색 목록으로 돌아가기 실패: {e}")
        return False

# 메인 크롤링 실행
for brand_name, query in search_queries:
    print(f"-------[+] {brand_name} 크롤링 시작!-------")
    
    try:
        # 네이버 지도 검색 URL 진입
        search_url = f"https://map.naver.com/p/search/{query}"
        driver.get(search_url)
        time.sleep(5)  # 페이지 로딩 대기
        
        print(f"-------{query} 검색 완료, 상세 페이지 로딩 중-------")
        
        # 검색어가 정확해서 바로 상세 페이지가 뜨는 경우
        # entryIframe으로 바로 전환 시도
        try:
            print("상세 페이지 iframe 찾는 중...")
            entry_iframe = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "iframe#entryIframe")))
            driver.switch_to.frame(entry_iframe)
            time.sleep(3)
            print("상세 페이지 iframe 전환 완료")
            
            # 기본 정보 수집
            restaurant_info = get_restaurant_basic_info()
            
            # 메뉴 탭 클릭 및 메뉴 정보 수집
            print('-------메뉴 탭 찾는중-------')
            if find_and_click_menu_tab():
                print("메뉴 정보 수집 시작...")
                menu_data = get_menu_list(driver, brand_name)
                all_menu_data[brand_name] = {
                    'restaurant_info': restaurant_info,
                    'menu_data': menu_data
                }
                print(f"[+] {brand_name} 메뉴 {len(menu_data)}개 수집 완료")
            else:
                print(f"[!] {brand_name} 메뉴 탭을 찾을 수 없습니다.")
                all_menu_data[brand_name] = {
                    'restaurant_info': restaurant_info,
                    'menu_data': []
                }
                
        except Exception as iframe_error:
            print(f"상세 페이지 iframe 접근 실패: {iframe_error}")
            # 혹시 검색 결과 페이지인 경우를 위한 대체 로직
            try:
                print("검색 결과 iframe 확인 중...")
                search_iframe = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "iframe#searchIframe")))
                driver.switch_to.frame(search_iframe)
                time.sleep(3)
                
                # 첫 번째 검색 결과 클릭
                selectors = [
                    "span.place_bluelink",
                    "a.place_bluelink", 
                    ".place_bluelink",
                    "span[class*='place_bluelink']",
                    "a[class*='place_bluelink']"
                ]
                
                first_result = None
                for selector in selectors:
                    try:
                        results = driver.find_elements(By.CSS_SELECTOR, selector)
                        if results:
                            first_result = results[0]
                            print(f"[+] 검색 결과 찾음: {selector}")
                            break
                    except:
                        continue
                
                if first_result and click_restaurant_from_list(first_result, 1):
                    # 상세 정보 iframe으로 전환
                    if switch_to_detail_iframe():
                        restaurant_info = get_restaurant_basic_info()
                        
                        if find_and_click_menu_tab():
                            menu_data = get_menu_list(driver, brand_name)
                            all_menu_data[brand_name] = {
                                'restaurant_info': restaurant_info,
                                'menu_data': menu_data
                            }
                            print(f"[+] {brand_name} 메뉴 {len(menu_data)}개 수집 완료")
                        else:
                            all_menu_data[brand_name] = {
                                'restaurant_info': restaurant_info,
                                'menu_data': []
                            }
                    else:
                        all_menu_data[brand_name] = []
                else:
                    all_menu_data[brand_name] = []
            except:
                all_menu_data[brand_name] = []
            
    except Exception as e:
        print(f"[!] {brand_name} 크롤링 실패: {e}")
        all_menu_data[brand_name] = []
    
    # 다음 검색을 위해 준비
    print("다음 검색 준비 중...")
    time.sleep(3)

# 최종 결과 정리
print("\n=== 최종 결과 ===")
for brand_name, data in all_menu_data.items():
    if isinstance(data, dict):
        menu_count = len(data.get('menu_data', []))
        print(f"{brand_name}: {menu_count}개 메뉴")
    else:
        print(f"{brand_name}: 0개 메뉴 (수집 실패)")

# 각 브랜드별로 리스트 분리
for brand_name, data in all_menu_data.items():
    if isinstance(data, dict) and data.get('menu_data'):
        print(f"\n--- {brand_name} 리스트 ---")
        menu_list = [item['menu'] for item in data['menu_data']]
        price_list = [item['price'] for item in data['menu_data']]
        desc_list = [item['description'] for item in data['menu_data']]
        
        print(f"메뉴: {menu_list}")
        print(f"가격: {price_list}")
        print(f"설명: {desc_list}")

# 결과를 JSON 파일로 저장
with open(os.path.join(BASE_PATH, "menu_data.json"), 'w', encoding='utf-8') as f:
    json.dump(all_menu_data, f, ensure_ascii=False, indent=2)

print("\n결과가 menu_data.json 파일로 저장되었습니다.")

# 브라우저 종료
# driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.action_chains import ActionChains
from selenium.common.exceptions import TimeoutException, NoSuchElementException
import time
import json
import os
import urllib.request
import re

# === 설정 ===
SCROLL_PAUSE_TIME = 2
BASE_PATH = r"C:\ai_x\source\Project\0709_데이터 폴더"

# 함수 설정
def safe_filename(name):
    return re.sub(r'[\\/*?:"<>|]', "", name)

def download_image(img_url, save_path):
    try:
        headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36',
            'Referer': 'https://map.naver.com/'
        }
        urllib.request.urlopen(urllib.request.Request(img_url, headers=headers))
        urllib.request.urlretrieve(img_url, save_path)
        return save_path
    except Exception as e:
        print(f"이미지 다운로드 실패: {e}")
        return None

# 10개 가게에 대한 브랜드명 리스트 
search_queries = [
    ("쉐이크쉑","쉐이크쉑 신림점"),
    ("맥도날드","맥도날드 신림점"),
    ("롯데리아","롯데리아 신림역점"),
    ("버거킹","버거킹 신림역점"),
    ("버거운버거","버거운버거 서울대점"),
    ("KFC","KFC 신림역"),
    ("노브랜드버거","노브랜드버거 신림남부점"),
    ("프랭크버거","프랭크버거 신림녹두거리점"),
    ("움버거앤윙스","움버거앤윙스 신림역점"),
    ("버거리","버거리 낙성대점")
]   

# 크롬 드라이버 설정
options = webdriver.ChromeOptions()
options.add_argument('user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36')
options.add_argument('window-size=1380,900')
options.add_argument('--no-sandbox')
options.add_argument('--disable-dev-shm-usage')
options.add_argument('--disable-blink-features=AutomationControlled')
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)

driver = webdriver.Chrome(options=options)
driver.execute_script("Object.defineProperty(navigator, 'webdriver', {get: () => undefined})")
wait = WebDriverWait(driver, 15)

# 전체 결과 저장용 딕셔너리
all_menu_data = {}

def click_restaurant_from_list(restaurant_element, index):
    """가게 클릭하여 세부 정보 패널 열기"""
    try:
        # 가게 링크 찾기 - 여러 셀렉터 시도
        link_selectors = [
            "a.place_bluelink",
            "span.place_bluelink", 
            ".place_bluelink",
            "a[class*='place_bluelink']",
            "span[class*='place_bluelink']"
        ]
        
        link_element = None
        for selector in link_selectors:
            try:
                link_element = restaurant_element.find_element(By.CSS_SELECTOR, selector)
                if link_element:
                    break
            except:
                continue
        
        if not link_element:
            print(f"가게 {index} 링크를 찾을 수 없습니다.")
            return False
        
        # 스크롤하여 요소가 보이도록 함
        driver.execute_script("arguments[0].scrollIntoView(true);", link_element)
        time.sleep(1)
        
        # 클릭
        ActionChains(driver).move_to_element(link_element).click().perform()
        print(f"가게 {index} 클릭 완료")
        
        # 세부 정보 패널 로딩 대기
        time.sleep(5)
        return True
        
    except Exception as e:
        print(f"가게 {index} 클릭 실패: {e}")
        return False

def switch_to_detail_iframe():
    """오른쪽 세부 정보 iframe으로 전환"""
    try:
        # 메인으로 돌아가기
        driver.switch_to.default_content()
        time.sleep(2)
        
        # 세부 정보 iframe 찾기
        detail_iframe = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "iframe#entryIframe")))
        driver.switch_to.frame(detail_iframe)
        print("세부 정보 iframe으로 전환 완료")
        time.sleep(3)
        return True
        
    except Exception as e:
        print(f"세부 정보 iframe 전환 실패: {e}")
        return False

def get_restaurant_basic_info():
    """가게 기본 정보 수집"""
    restaurant_info = {}
    
    try:
        # 가게 이름
        name_selectors = ["span.GHAhO", "h1", ".name", "[class*='name']"]
        for selector in name_selectors:
            try:
                name_element = driver.find_element(By.CSS_SELECTOR, selector)
                restaurant_info['name'] = name_element.text.strip()
                print(f"가게명: {restaurant_info['name']}")
                break
            except:
                continue
        
        if 'name' not in restaurant_info:
            restaurant_info['name'] = "가게명 정보 없음"
        
        # 가게 분류
        try:
            category_element = driver.find_element(By.CSS_SELECTOR, "span.lnJFt")
            restaurant_info['category'] = category_element.text.strip()
            print(f"분류: {restaurant_info['category']}")
        except:
            restaurant_info['category'] = "분류 정보 없음"
        
        # 가게 주소
        try:
            address_element = driver.find_element(By.CSS_SELECTOR, "span.LDgIH")
            restaurant_info['address'] = address_element.text.strip()
            print(f"주소: {restaurant_info['address']}")
        except:
            restaurant_info['address'] = "주소 정보 없음"
            
    except Exception as e:
        print(f"기본 정보 수집 실패: {e}")
        restaurant_info = {'name': "정보 수집 실패", 'category': "", 'address': ""}
        
    return restaurant_info

def find_and_click_menu_tab():
    """메뉴 탭을 정확하게 찾아서 클릭 - 모든 탭 확인"""
    try:
        print("메뉴 탭 찾는 중...")
        
        # 페이지 스크롤하여 탭이 보이도록 함
        driver.execute_script("window.scrollTo(0, 0);")
        time.sleep(1)
        
        # 1단계: 실제 메뉴 탭 셀렉터 사용
        tab_selectors = [
            "a.tpj9w._tab-menu",  # 사용자가 제공한 정확한 클래스
            "a._tab-menu",        # _tab-menu 클래스만으로도 시도
            "a.tpj9w",            # tpj9w 클래스만으로도 시도
            "div.YYh8o a",
            "div[class*='tab'] a",
            "nav a",
            "ul li a"
        ]
        
        all_tabs = []
        for selector in tab_selectors:
            try:
                tabs = driver.find_elements(By.CSS_SELECTOR, selector)
                if tabs:
                    all_tabs = tabs
                    print(f"탭 요소들 찾음: {selector} - {len(tabs)}개")
                    break
            except:
                continue
        
        if not all_tabs:
            print("탭 요소를 찾을 수 없습니다")
            return False
        
        # 2단계: 모든 탭의 정보를 출력하고 메뉴 탭 찾기
        print("=== 모든 탭 정보 확인 ===")
        menu_tab = None
        for i, tab in enumerate(all_tabs):
            try:
                # 탭 텍스트 확인
                tab_text = tab.text.strip()
                # href 속성 확인
                tab_href = tab.get_attribute("href") or ""
                # class 속성 확인
                tab_class = tab.get_attribute("class") or ""
                
                print(f"탭 {i+1}: 텍스트='{tab_text}', class='{tab_class}', href='{tab_href[-50:] if len(tab_href) > 50 else tab_href}'")
                
                # 메뉴 탭인지 판단 - 정확한 조건 적용
                is_menu_tab = (
                    "메뉴" in tab_text or
                    "menu" in tab_text.lower() or
                    "menu" in tab_href.lower()
                )
                
                if is_menu_tab:
                    menu_tab = tab
                    print(f"*** 메뉴 탭 발견: '{tab_text}' (class: {tab_class}) ***")
                    break
                    
            except Exception as e:
                print(f"탭 {i+1} 확인 중 오류: {e}")
                continue
        
        print("=== 탭 정보 확인 완료 ===")
        
        # 3단계: 메뉴 탭 클릭
        if menu_tab:
            try:
                # 스크롤하여 요소가 보이도록 함
                driver.execute_script("arguments[0].scrollIntoView(true);", menu_tab)
                time.sleep(1)
                
                # JavaScript로 클릭
                driver.execute_script("arguments[0].click();", menu_tab)
                time.sleep(5)  # 메뉴 로딩 충분히 기다리기
                print("메뉴 탭 클릭 완료")
                return True
                
            except Exception as e:
                print(f"메뉴 탭 클릭 실패: {e}")
                return False
        else:
            print("메뉴 탭을 찾을 수 없습니다")
            # 메뉴 탭을 찾을 수 없을 때 모든 탭을 한번 더 클릭해보기
            print("모든 탭을 순차적으로 확인해보겠습니다...")
            for i, tab in enumerate(all_tabs):
                try:
                    tab_text = tab.text.strip()
                    print(f"탭 {i+1} '{tab_text}' 클릭해보기...")
                    driver.execute_script("arguments[0].click();", tab)
                    time.sleep(3)
                    
                    # 메뉴 요소가 있는지 확인
                    menu_elements = driver.find_elements(By.CSS_SELECTOR, "li.E2jtL")
                    if menu_elements:
                        print(f"탭 {i+1} '{tab_text}'에서 메뉴 요소 {len(menu_elements)}개 발견!")
                        return True
                except Exception as e:
                    print(f"탭 {i+1} 클릭 실패: {e}")
                    continue
            return False
            
    except Exception as e:
        print(f"메뉴 탭 찾기 실패: {e}")
        return False

def get_menu_list(driver, brand_name):
    """메뉴 탭에서 메뉴 정보 수집 - order_list_inner 반복 구조에 맞게 수정"""
    menu_data = []

    try:
        print("메뉴 페이지 로딩 대기 중...")
        time.sleep(3)
        
        # 메뉴 더보기 버튼 클릭 (있다면)
        try:
            more_button_selectors = [
                "a.fvwqf",
                "button.fvwqf",
                "a[class*='more']",
                "button[class*='more']",
                "//a[contains(text(), '더보기')]",
                "//button[contains(text(), '더보기')]"
            ]
            
            for selector in more_button_selectors:
                try:
                    if selector.startswith("//"):
                        more_button = driver.find_element(By.XPATH, selector)
                    else:
                        more_button = driver.find_element(By.CSS_SELECTOR, selector)
                    
                    if more_button.is_displayed() and more_button.is_enabled():
                        driver.execute_script("arguments[0].scrollIntoView(true);", more_button)
                        time.sleep(1)
                        driver.execute_script("arguments[0].click();", more_button)
                        time.sleep(2)
                        print("메뉴 더보기 버튼 클릭 완료")
                        break
                except:
                    continue
        except:
            print("메뉴 더보기 버튼이 없거나 이미 모든 메뉴가 표시됨")
        
        # 모든 order_list_inner 찾기
        order_list_inners = driver.find_elements(By.CSS_SELECTOR, "div.order_list_inner")
        print(f"order_list_inner 섹션 {len(order_list_inners)}개 발견")
        
        for section_idx, order_inner in enumerate(order_list_inners):
            try:
                print(f"\n=== 섹션 {section_idx + 1} 처리 시작 ===")
                
                # 현재 섹션의 카테고리 제목 찾기 (order_list_tit가 있다면)
                category_title = "기타"
                try:
                    # order_list_inner 이전이나 내부에 있는 카테고리 제목 찾기
                    category_element = order_inner.find_element(By.XPATH, "./preceding-sibling::div[contains(@class, 'order_list_tit')]//span[@class='title']")
                    category_title = category_element.text.strip()
                except:
                    try:
                        # order_list_inner 내부에 카테고리 제목이 있는지 확인
                        category_element = order_inner.find_element(By.CSS_SELECTOR, "div.order_list_tit span.title")
                        category_title = category_element.text.strip()
                    except:
                        # 섹션 인덱스로 카테고리 추정
                        category_titles = ["햄버거", "사이드", "쉐이크", "음료", "굿즈"]
                        if section_idx < len(category_titles):
                            category_title = category_titles[section_idx]
                        else:
                            category_title = f"카테고리_{section_idx + 1}"
                
                print(f"섹션 {section_idx + 1} 카테고리: {category_title}")
                
                # 음료, 굿즈 카테고리는 건너뛰기
                if category_title in ["음료", "굿즈"]:
                    print(f"카테고리 '{category_title}' 건너뛰기")
                    continue
                
                # 현재 order_list_inner 내의 ul과 li 찾기
                try:
                    ul_element = order_inner.find_element(By.CSS_SELECTOR, "ul")
                    li_elements = ul_element.find_elements(By.CSS_SELECTOR, "li")
                    print(f"섹션 {section_idx + 1}에서 li 요소 {len(li_elements)}개 발견")
                except:
                    print(f"섹션 {section_idx + 1}에서 ul/li 요소를 찾을 수 없음")
                    continue
                
                # 각 li 요소에서 메뉴 정보 추출
                for li_idx, li_element in enumerate(li_elements):
                    try:
                        menu_info = {}
                        
                        # 메뉴명 추출
                        try:
                            name_element = li_element.find_element(By.CSS_SELECTOR, "div.MenuContent__tit__313LA")
                            menu_info['menu'] = name_element.text.strip()
                        except:
                            print(f"  li {li_idx + 1}: 메뉴명 없음, 건너뛰기")
                            continue
                        
                        # 메뉴 설명 추출
                        try:
                            desc_container = li_element.find_element(By.CSS_SELECTOR, "div.MenuContent__detail__OG635")
                            desc_element = desc_container.find_element(By.CSS_SELECTOR, "span.detail_txt")
                            menu_info['description'] = desc_element.text.strip()
                        except:
                            menu_info['description'] = "설명 없음"
                        
                        # 가격 추출 (strong 태그 안에 있음)
                        try:
                            price_container = li_element.find_element(By.CSS_SELECTOR, "div.MenuContent__price__lhCy9")
                            price_element = price_container.find_element(By.CSS_SELECTOR, "strong")
                            menu_info['price'] = price_element.text.strip()
                        except:
                            menu_info['price'] = "가격 정보 없음"
                        
                        # 이미지 추출
                        try:
                            img_container = li_element.find_element(By.CSS_SELECTOR, "div.MenuContent__info_img__JnyVE")
                            img_element = img_container.find_element(By.CSS_SELECTOR, "img[alt='섬네일']")
                            img_url = img_element.get_attribute("src")
                            menu_info['image_url'] = img_url
                            
                            if img_url:
                                # 이미지 저장
                                safe_menu_name = safe_filename(menu_info['menu'])[:30]
                                menu_folder = os.path.join(BASE_PATH, brand_name, safe_menu_name)
                                os.makedirs(menu_folder, exist_ok=True)
                                
                                img_filename = f"brand_{len(menu_data)+1:03d}.jpg"
                                save_path = os.path.join(menu_folder, img_filename)
                                
                                image_path = download_image(img_url, save_path)
                                menu_info['image_path'] = image_path
                            else:
                                menu_info['image_path'] = None
                        except:
                            menu_info['image_url'] = None
                            menu_info['image_path'] = None
                        
                        # 카테고리 정보 추가
                        menu_info['category'] = category_title
                        
                        menu_data.append(menu_info)
                        print(f"  [{category_title}] {len(menu_data)}. {menu_info['menu']} - {menu_info['price']}")
                        
                    except Exception as e:
                        print(f"  li {li_idx + 1} 처리 중 오류: {e}")
                        continue
                        
                print(f"섹션 {section_idx + 1} 완료 - 총 {len([m for m in menu_data if m['category'] == category_title])}개 메뉴 수집")
                        
            except Exception as e:
                print(f"섹션 {section_idx + 1} 처리 중 오류: {e}")
                continue
        
        # 대체 방법: order_list_inner를 찾을 수 없는 경우
        if len(menu_data) == 0:
            print("\n대체 방법으로 메뉴 수집 시도...")
            
            # 메뉴명이 있는 모든 요소 직접 찾기
            menu_name_elements = driver.find_elements(By.CSS_SELECTOR, "div.MenuContent__tit__313LA")
            print(f"메뉴명 요소 {len(menu_name_elements)}개 발견")
            
            for idx, name_element in enumerate(menu_name_elements):
                try:
                    menu_info = {}
                    
                    # 메뉴명
                    menu_info['menu'] = name_element.text.strip()
                    
                    # 부모 li 요소 찾기
                    try:
                        li_element = name_element.find_element(By.XPATH, "./ancestor::li[1]")
                    except:
                        print(f"메뉴 {idx + 1}: 부모 li 요소를 찾을 수 없음")
                        continue
                    
                    # 메뉴 설명
                    try:
                        desc_container = li_element.find_element(By.CSS_SELECTOR, "div.MenuContent__detail__OG635")
                        desc_element = desc_container.find_element(By.CSS_SELECTOR, "span.detail_txt")
                        menu_info['description'] = desc_element.text.strip()
                    except:
                        menu_info['description'] = "설명 없음"
                    
                    # 가격
                    try:
                        price_container = li_element.find_element(By.CSS_SELECTOR, "div.MenuContent__price__lhCy9")
                        price_element = price_container.find_element(By.CSS_SELECTOR, "strong")
                        menu_info['price'] = price_element.text.strip()
                    except:
                        menu_info['price'] = "가격 정보 없음"
                    
                    # 이미지
                    try:
                        img_container = li_element.find_element(By.CSS_SELECTOR, "div.MenuContent__info_img__JnyVE")
                        img_element = img_container.find_element(By.CSS_SELECTOR, "img[alt='섬네일']")
                        img_url = img_element.get_attribute("src")
                        menu_info['image_url'] = img_url
                        
                        if img_url:
                            safe_menu_name = safe_filename(menu_info['menu'])[:30]
                            menu_folder = os.path.join(BASE_PATH, brand_name, safe_menu_name)
                            os.makedirs(menu_folder, exist_ok=True)
                            
                            img_filename = f"brand_{len(menu_data)+1:03d}.jpg"
                            save_path = os.path.join(menu_folder, img_filename)
                            
                            image_path = download_image(img_url, save_path)
                            menu_info['image_path'] = image_path
                        else:
                            menu_info['image_path'] = None
                    except:
                        menu_info['image_url'] = None
                        menu_info['image_path'] = None
                    
                    # 카테고리는 인덱스로 추정
                    category_titles = ["햄버거", "사이드", "쉐이크"]
                    menu_info['category'] = category_titles[idx % len(category_titles)]
                    
                    menu_data.append(menu_info)
                    print(f"  [{menu_info['category']}] {len(menu_data)}. {menu_info['menu']} - {menu_info['price']}")
                    
                except Exception as e:
                    print(f"메뉴 {idx + 1} 처리 중 오류: {e}")
                    continue
                
    except Exception as e:
        print(f"메뉴 정보 수집 실패: {e}")
    
    print(f"\n총 {len(menu_data)}개 메뉴 수집 완료")
    return menu_data

def go_back_to_search_list():
    """검색 목록으로 돌아가기"""
    try:
        # 메인으로 돌아가기
        driver.switch_to.default_content()
        time.sleep(2)
        
        # 검색 결과 iframe으로 다시 전환
        search_iframe = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "iframe#searchIframe")))
        driver.switch_to.frame(search_iframe)
        time.sleep(2)
        print("검색 목록으로 돌아가기 완료")
        return True
        
    except Exception as e:
        print(f"검색 목록으로 돌아가기 실패: {e}")
        return False

# 메인 크롤링 실행
for brand_name, query in search_queries:
    print(f"-------[+] {brand_name} 크롤링 시작!-------")
    
    try:
        # 네이버 지도 검색 URL 진입
        search_url = f"https://map.naver.com/p/search/{query}"
        driver.get(search_url)
        time.sleep(5)  # 페이지 로딩 대기
        
        print(f"-------{query} 검색 완료, 상세 페이지 로딩 중-------")
        
        # 검색어가 정확해서 바로 상세 페이지가 뜨는 경우
        # entryIframe으로 바로 전환 시도
        try:
            print("상세 페이지 iframe 찾는 중...")
            entry_iframe = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "iframe#entryIframe")))
            driver.switch_to.frame(entry_iframe)
            time.sleep(3)
            print("상세 페이지 iframe 전환 완료")
            
            # 기본 정보 수집
            restaurant_info = get_restaurant_basic_info()
            
            # 메뉴 탭 클릭 및 메뉴 정보 수집
            print('-------메뉴 탭 찾는중-------')
            if find_and_click_menu_tab():
                print("메뉴 정보 수집 시작...")
                menu_data = get_menu_list(driver, brand_name)
                all_menu_data[brand_name] = {
                    'restaurant_info': restaurant_info,
                    'menu_data': menu_data
                }
                print(f"[+] {brand_name} 메뉴 {len(menu_data)}개 수집 완료")
            else:
                print(f"[!] {brand_name} 메뉴 탭을 찾을 수 없습니다.")
                all_menu_data[brand_name] = {
                    'restaurant_info': restaurant_info,
                    'menu_data': []
                }
                
        except Exception as iframe_error:
            print(f"상세 페이지 iframe 접근 실패: {iframe_error}")
            # 혹시 검색 결과 페이지인 경우를 위한 대체 로직
            try:
                print("검색 결과 iframe 확인 중...")
                search_iframe = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "iframe#searchIframe")))
                driver.switch_to.frame(search_iframe)
                time.sleep(3)
                
                # 첫 번째 검색 결과 클릭
                selectors = [
                    "span.place_bluelink",
                    "a.place_bluelink", 
                    ".place_bluelink",
                    "span[class*='place_bluelink']",
                    "a[class*='place_bluelink']"
                ]
                
                first_result = None
                for selector in selectors:
                    try:
                        results = driver.find_elements(By.CSS_SELECTOR, selector)
                        if results:
                            first_result = results[0]
                            print(f"[+] 검색 결과 찾음: {selector}")
                            break
                    except:
                        continue
                
                if first_result and click_restaurant_from_list(first_result, 1):
                    # 상세 정보 iframe으로 전환
                    if switch_to_detail_iframe():
                        restaurant_info = get_restaurant_basic_info()
                        
                        if find_and_click_menu_tab():
                            menu_data = get_menu_list(driver, brand_name)
                            all_menu_data[brand_name] = {
                                'restaurant_info': restaurant_info,
                                'menu_data': menu_data
                            }
                            print(f"[+] {brand_name} 메뉴 {len(menu_data)}개 수집 완료")
                        else:
                            all_menu_data[brand_name] = {
                                'restaurant_info': restaurant_info,
                                'menu_data': []
                            }
                    else:
                        all_menu_data[brand_name] = []
                else:
                    all_menu_data[brand_name] = []
            except:
                all_menu_data[brand_name] = []
            
    except Exception as e:
        print(f"[!] {brand_name} 크롤링 실패: {e}")
        all_menu_data[brand_name] = []
    
    # 다음 검색을 위해 준비
    print("다음 검색 준비 중...")
    time.sleep(3)

# 최종 결과 정리
print("\n=== 최종 결과 ===")
for brand_name, data in all_menu_data.items():
    if isinstance(data, dict):
        menu_count = len(data.get('menu_data', []))
        print(f"{brand_name}: {menu_count}개 메뉴")
    else:
        print(f"{brand_name}: 0개 메뉴 (수집 실패)")

# 각 브랜드별로 리스트 분리
for brand_name, data in all_menu_data.items():
    if isinstance(data, dict) and data.get('menu_data'):
        print(f"\n--- {brand_name} 리스트 ---")
        menu_list = [item['menu'] for item in data['menu_data']]
        price_list = [item['price'] for item in data['menu_data']]
        desc_list = [item['description'] for item in data['menu_data']]
        
        print(f"메뉴: {menu_list}")
        print(f"가격: {price_list}")
        print(f"설명: {desc_list}")

# 결과를 JSON 파일로 저장
with open(os.path.join(BASE_PATH, "menu_data.json"), 'w', encoding='utf-8') as f:
    json.dump(all_menu_data, f, ensure_ascii=False, indent=2)

print("\n결과가 menu_data.json 파일로 저장되었습니다.")

# 브라우저 종료
# driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.chrome.service import Service
from selenium.common.exceptions import TimeoutException, NoSuchElementException
import time
import json
import os
import urllib.request
import re

# === 설정 ===
BASE_PATH = r"C:\ai_x\source\Project\0709_데이터 폴더"

# 함수 설정
def safe_filename(name):
    return re.sub(r'[\\/*?:"<>|]', "", name)

def download_image(img_url, save_path):
    try:
        headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'
        }
        urllib.request.urlopen(urllib.request.Request(img_url, headers=headers))
        urllib.request.urlretrieve(img_url, save_path)
        return save_path
    except Exception as e:
        print(f"이미지 다운로드 실패: {e}")
        return None

# 10개 가게 리스트 
search_queries = [
    ("쉐이크쉑","쉐이크쉑 신림점"),
    ("맥도날드","맥도날드 신림점"),
    ("롯데리아","롯데리아 신림역점"),
    ("버거킹","버거킹 신림역점"),
    ("버거운버거","버거운버거 서울대점"),
    ("KFC","KFC 신림역"),
    ("노브랜드버거","노브랜드버거 신림남부점"),
    ("프랭크버거","프랭크버거 신림녹두거리점"),
    ("움버거앤윙스","움버거앤윙스 신림역점"),
    ("버거리","버거리 낙성대점")
]   

# 크롬 드라이버 설정
service = Service(port=9999)
options = webdriver.ChromeOptions()
options.add_argument('user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36')
options.add_argument('window-size=1380,900')
options.add_argument('--no-sandbox')
options.add_argument('--disable-dev-shm-usage')
options.add_experimental_option("excludeSwitches", ["enable-automation"])

driver = webdriver.Chrome(service=service, options=options)
wait = WebDriverWait(driver, 15)
all_menu_data = {}

def get_restaurant_basic_info():
    """가게 기본 정보 수집"""
    restaurant_info = {}
    
    try:
        # 가게 이름
        name_selectors = ["span.GHAhO", "h1", "[class*='name']"]
        for selector in name_selectors:
            try:
                name_element = driver.find_element(By.CSS_SELECTOR, selector)
                restaurant_info['name'] = name_element.text.strip()
                print(f"가게명: {restaurant_info['name']}")
                break
            except:
                continue
        
        # 기본값 설정
        if 'name' not in restaurant_info:
            restaurant_info['name'] = "가게명 정보 없음"
        
        # 가게 분류
        try:
            category_element = driver.find_element(By.CSS_SELECTOR, "span.lnJFt")
            restaurant_info['category'] = category_element.text.strip()
        except:
            restaurant_info['category'] = "분류 정보 없음"
        
        # 가게 주소
        try:
            address_element = driver.find_element(By.CSS_SELECTOR, "span.LDgIH")
            restaurant_info['address'] = address_element.text.strip()
        except:
            restaurant_info['address'] = "주소 정보 없음"
            
    except Exception as e:
        print(f"기본 정보 수집 실패: {e}")
        restaurant_info = {'name': "정보 수집 실패", 'category': "", 'address': ""}
        
    return restaurant_info

def find_and_click_menu_tab():
    """메뉴 탭 찾아서 클릭"""
    try:
        print("메뉴 탭 찾는 중...")
        driver.execute_script("window.scrollTo(0, 0);")
        time.sleep(1)
        
        # 탭 요소들 찾기
        tab_selectors = [
            "a.tpj9w._tab-menu",
            "a._tab-menu",
            "a.tpj9w",
            "div.YYh8o a",
            "div[class*='tab'] a",
            "nav a",
            "ul li a"
        ]
        
        all_tabs = []
        for selector in tab_selectors:
            try:
                tabs = driver.find_elements(By.CSS_SELECTOR, selector)
                if tabs:
                    all_tabs = tabs
                    print(f"탭 요소들 찾음: {len(tabs)}개")
                    break
            except:
                continue
        
        if not all_tabs:
            print("탭 요소를 찾을 수 없습니다")
            return False
        
        # 메뉴 탭 찾기
        menu_tab = None
        for i, tab in enumerate(all_tabs):
            try:
                tab_text = tab.text.strip()
                tab_href = tab.get_attribute("href") or ""
                
                is_menu_tab = (
                    "메뉴" in tab_text or
                    "menu" in tab_text.lower() or
                    "menu" in tab_href.lower()
                )
                
                if is_menu_tab:
                    menu_tab = tab
                    print(f"메뉴 탭 발견: '{tab_text}'")
                    break
            except:
                continue
        
        # 메뉴 탭 클릭
        if menu_tab:
            driver.execute_script("arguments[0].click();", menu_tab)
            time.sleep(5)
            print("메뉴 탭 클릭 완료")
            return True
        else:
            print("메뉴 탭을 찾을 수 없습니다. 모든 탭을 순차적으로 확인...")
            # 모든 탭 순차 클릭해서 메뉴 찾기
            for i, tab in enumerate(all_tabs):
                try:
                    tab_text = tab.text.strip()
                    print(f"탭 {i+1} '{tab_text}' 클릭해보기...")
                    driver.execute_script("arguments[0].click();", tab)
                    time.sleep(3)
                    
                    # 메뉴 요소 존재 확인
                    menu_elements = driver.find_elements(By.CSS_SELECTOR, "li")
                    for elem in menu_elements:
                        try:
                            text = elem.text.strip()
                            if text and ("원" in text or "," in text) and any(char.isdigit() for char in text):
                                print(f"탭 {i+1} '{tab_text}'에서 메뉴 발견!")
                                return True
                        except:
                            continue
                except:
                    continue
            return False
            
    except Exception as e:
        print(f"메뉴 탭 찾기 실패: {e}")
        return False

def get_menu_list(driver, brand_name):
    """메뉴 정보 수집 - 3단계 방법"""
    menu_data = []
    
    try:
        print("메뉴 페이지 로딩 대기...")
        time.sleep(5)
        
        # 더보기 버튼 클릭
        more_selectors = ["a.fvwqf", "button.fvwqf", "//a[contains(text(), '더보기')]"]
        for selector in more_selectors:
            try:
                if selector.startswith("//"):
                    more_button = driver.find_element(By.XPATH, selector)
                else:
                    more_button = driver.find_element(By.CSS_SELECTOR, selector)
                
                if more_button.is_displayed():
                    driver.execute_script("arguments[0].click();", more_button)
                    time.sleep(3)
                    print("더보기 버튼 클릭 완료")
                    break
            except:
                continue
        
        # 방법 1: 일반적인 li 구조 확인
        try:
            print("방법 1: li 구조 확인...")
            menu_items = driver.find_elements(By.CSS_SELECTOR, "li")
            menu_candidates = []
            
            for item in menu_items:
                try:
                    text_elements = item.find_elements(By.CSS_SELECTOR, "*")
                    menu_name = None
                    price = None
                    description = None
                    
                    for elem in text_elements:
                        text = elem.text.strip()
                        if text and len(text) > 1:
                            if "원" in text or ("," in text and any(char.isdigit() for char in text)):
                                if price is None:
                                    price = text
                            elif len(text) > 2 and len(text) < 50 and not text.isdigit():
                                if menu_name is None:
                                    menu_name = text
                                elif len(text) > 10 and description is None:
                                    description = text
                    
                    if menu_name and price:
                        menu_candidates.append({
                            'element': item,
                            'menu': menu_name,
                            'price': price,
                            'description': description or "설명 없음"
                        })
                except:
                    continue
            
            if len(menu_candidates) > 2:
                print(f"방법 1 성공: {len(menu_candidates)}개 메뉴 발견")
                
                for candidate in menu_candidates:
                    menu_info = {
                        'menu': candidate['menu'],
                        'price': candidate['price'],
                        'description': candidate['description'],
                        'category': '메뉴'
                    }
                    
                    # 이미지 처리
                    try:
                        img_element = candidate['element'].find_element(By.CSS_SELECTOR, "img")
                        img_url = img_element.get_attribute("src")
                        menu_info['image_url'] = img_url
                        
                        if img_url:
                            safe_menu_name = safe_filename(menu_info['menu'])[:30]
                            menu_folder = os.path.join(BASE_PATH, brand_name, safe_menu_name)
                            os.makedirs(menu_folder, exist_ok=True)
                            
                            img_filename = f"brand_{len(menu_data)+1:03d}.jpg"
                            save_path = os.path.join(menu_folder, img_filename)
                            menu_info['image_path'] = download_image(img_url, save_path)
                        else:
                            menu_info['image_path'] = None
                    except:
                        menu_info['image_url'] = None
                        menu_info['image_path'] = None
                    
                    menu_data.append(menu_info)
                    print(f"  {len(menu_data)}. {menu_info['menu']} - {menu_info['price']}")
                
                return menu_data
        except Exception as e:
            print(f"방법 1 실패: {e}")
        
        # 방법 2: 패턴 기반 클래스명 찾기
        try:
            print("방법 2: 패턴 기반 클래스명 찾기...")
            
            patterns = ["*[class*='tit']", "*[class*='name']", "*[class*='menu']"]
            
            for pattern in patterns:
                try:
                    name_elements = driver.find_elements(By.CSS_SELECTOR, pattern)
                    if len(name_elements) > 3:
                        print(f"패턴 '{pattern}' 발견: {len(name_elements)}개")
                        
                        for idx, name_elem in enumerate(name_elements):
                            try:
                                menu_name = name_elem.text.strip()
                                if not menu_name or len(menu_name) < 2:
                                    continue
                                
                                parent = name_elem.find_element(By.XPATH, "./ancestor::li[1] | ./ancestor::div[1]")
                                
                                # 가격 찾기
                                price = "가격 정보 없음"
                                price_patterns = ["*[class*='price']", "strong", "em"]
                                for price_pattern in price_patterns:
                                    try:
                                        price_elem = parent.find_element(By.CSS_SELECTOR, price_pattern)
                                        price_text = price_elem.text.strip()
                                        if "원" in price_text or ("," in price_text and any(char.isdigit() for char in price_text)):
                                            price = price_text
                                            break
                                    except:
                                        continue
                                
                                menu_info = {
                                    'menu': menu_name,
                                    'price': price,
                                    'description': "설명 없음",
                                    'category': '메뉴'
                                }
                                
                                # 이미지 처리
                                try:
                                    img_elem = parent.find_element(By.CSS_SELECTOR, "img")
                                    img_url = img_elem.get_attribute("src")
                                    menu_info['image_url'] = img_url
                                    
                                    if img_url:
                                        safe_menu_name = safe_filename(menu_info['menu'])[:30]
                                        menu_folder = os.path.join(BASE_PATH, brand_name, safe_menu_name)
                                        os.makedirs(menu_folder, exist_ok=True)
                                        
                                        img_filename = f"brand_{len(menu_data)+1:03d}.jpg"
                                        save_path = os.path.join(menu_folder, img_filename)
                                        menu_info['image_path'] = download_image(img_url, save_path)
                                    else:
                                        menu_info['image_path'] = None
                                except:
                                    menu_info['image_url'] = None
                                    menu_info['image_path'] = None
                                
                                menu_data.append(menu_info)
                                print(f"  {len(menu_data)}. {menu_info['menu']} - {menu_info['price']}")
                                
                            except:
                                continue
                        
                        if len(menu_data) > 0:
                            return menu_data
                except:
                    continue
        except Exception as e:
            print(f"방법 2 실패: {e}")
        
        # 방법 3: 텍스트 기반 검색
        try:
            print("방법 3: 텍스트 기반 메뉴 검색...")
            
            all_elements = driver.find_elements(By.CSS_SELECTOR, "*")
            menu_candidates = []
            
            for elem in all_elements:
                try:
                    text = elem.text.strip()
                    if text and ("원" in text or ("," in text and any(char.isdigit() for char in text))):
                        parent = elem.find_element(By.XPATH, "./ancestor::*[position()<=3]")
                        
                        # 메뉴명 찾기
                        text_elements = parent.find_elements(By.CSS_SELECTOR, "*")
                        menu_name = None
                        
                        for text_elem in text_elements:
                            elem_text = text_elem.text.strip()
                            if (elem_text and len(elem_text) > 2 and len(elem_text) < 30 and 
                                not ("원" in elem_text or elem_text.isdigit()) and elem_text != text):
                                menu_name = elem_text
                                break
                        
                        if menu_name:
                            menu_candidates.append({
                                'menu': menu_name,
                                'price': text,
                                'description': "설명 없음",
                                'element': parent
                            })
                except:
                    continue
            
            # 중복 제거
            unique_menus = []
            for candidate in menu_candidates:
                if not any(existing['menu'] == candidate['menu'] for existing in unique_menus):
                    unique_menus.append(candidate)
            
            if len(unique_menus) > 2:
                print(f"방법 3 성공: {len(unique_menus)}개 메뉴 발견")
                
                for candidate in unique_menus:
                    menu_info = {
                        'menu': candidate['menu'],
                        'price': candidate['price'],
                        'description': candidate['description'],
                        'category': '메뉴'
                    }
                    
                    # 이미지 처리
                    try:
                        img_elem = candidate['element'].find_element(By.CSS_SELECTOR, "img")
                        img_url = img_elem.get_attribute("src")
                        menu_info['image_url'] = img_url
                        
                        if img_url:
                            safe_menu_name = safe_filename(menu_info['menu'])[:30]
                            menu_folder = os.path.join(BASE_PATH, brand_name, safe_menu_name)
                            os.makedirs(menu_folder, exist_ok=True)
                            
                            img_filename = f"brand_{len(menu_data)+1:03d}.jpg"
                            save_path = os.path.join(menu_folder, img_filename)
                            menu_info['image_path'] = download_image(img_url, save_path)
                        else:
                            menu_info['image_path'] = None
                    except:
                        menu_info['image_url'] = None
                        menu_info['image_path'] = None
                    
                    menu_data.append(menu_info)
                    print(f"  {len(menu_data)}. {menu_info['menu']} - {menu_info['price']}")
                
                return menu_data
        except Exception as e:
            print(f"방법 3 실패: {e}")
        
        print("모든 방법 실패: 메뉴를 찾을 수 없습니다")
        
    except Exception as e:
        print(f"메뉴 정보 수집 실패: {e}")
    
    print(f"총 {len(menu_data)}개 메뉴 수집 완료")
    return menu_data

# 메인 크롤링 실행
try:
    for brand_name, query in search_queries:
        print(f"\n-------[+] {brand_name} 크롤링 시작!-------")
        
        try:
            # 검색 URL 접속
            search_url = f"https://map.naver.com/p/search/{query}"
            driver.get(search_url)
            time.sleep(5)
            
            print(f"검색 완료: {query}")
            
            # 상세 페이지 iframe 접근
            try:
                entry_iframe = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "iframe#entryIframe")))
                driver.switch_to.frame(entry_iframe)
                time.sleep(3)
                print("상세 페이지 iframe 전환 완료")
                
                # 기본 정보 수집
                restaurant_info = get_restaurant_basic_info()
                
                # 메뉴 탭 클릭 및 메뉴 정보 수집
                if find_and_click_menu_tab():
                    menu_data = get_menu_list(driver, brand_name)
                    all_menu_data[brand_name] = {
                        'restaurant_info': restaurant_info,
                        'menu_data': menu_data
                    }
                    print(f"[+] {brand_name} 메뉴 {len(menu_data)}개 수집 완료")
                else:
                    all_menu_data[brand_name] = {
                        'restaurant_info': restaurant_info,
                        'menu_data': []
                    }
                    
            except Exception as iframe_error:
                print(f"상세 페이지 접근 실패: {iframe_error}")
                all_menu_data[brand_name] = []
                
        except Exception as e:
            print(f"[!] {brand_name} 크롤링 실패: {e}")
            all_menu_data[brand_name] = []
        
        print("다음 검색 준비 중...")
        time.sleep(3)

    # 최종 결과 출력
    print("\n=== 최종 결과 ===")
    for brand_name, data in all_menu_data.items():
        if isinstance(data, dict):
            menu_count = len(data.get('menu_data', []))
            print(f"{brand_name}: {menu_count}개 메뉴")
        else:
            print(f"{brand_name}: 0개 메뉴 (수집 실패)")

    # 브랜드별 리스트 출력
    for brand_name, data in all_menu_data.items():
        if isinstance(data, dict) and data.get('menu_data'):
            print(f"\n--- {brand_name} 리스트 ---")
            menu_list = [item['menu'] for item in data['menu_data']]
            price_list = [item['price'] for item in data['menu_data']]
            desc_list = [item['description'] for item in data['menu_data']]
            
            print(f"메뉴: {menu_list}")
            print(f"가격: {price_list}")
            print(f"설명: {desc_list}")

    # JSON 파일 저장
    with open(os.path.join(BASE_PATH, "menu_data.json"), 'w', encoding='utf-8') as f:
        json.dump(all_menu_data, f, ensure_ascii=False, indent=2)

    print("\n결과가 menu_data.json 파일로 저장되었습니다.")

except Exception as e:
    print(f"전체 프로세스 오류: {e}")

finally:
    # 브라우저 종료
    try:
        pass
#         driver.quit()
#         print("브라우저 종료 완료")
    except:
        pass
```

```python
# 2번 테스트 
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

search_queries = [
    ("쉐이크쉑","쉐이크쉑 신림점"),
    ("맥도날드","맥도날드 신림점"),
    ("롯데리아","롯데리아 신림역점"),
    ("버거킹","버거킹 신림역점"),
    ("버거운버거","버거운버거 서울대점"),
    ("KFC","KFC 신림역"),
    ("노브랜드버거","노브랜드버거 신림남부점"),
    ("프랭크버거","프랭크버거 신림녹두거리점"),
    ("움버거앤윙스","움버거앤윙스 신림역점"),
    ("버거리","버거리 낙성대점")
]

options = webdriver.ChromeOptions()
options.add_argument("--start-maximized")
driver = webdriver.Chrome(options=options)
wait = WebDriverWait(driver, 10)

def click_more_buttons():
    """더보기 버튼 모두 클릭"""
    while True:
        try:
            more_btn = driver.find_element(By.CSS_SELECTOR, ".fvwqf[role='button']")
            if more_btn.is_displayed():
                driver.execute_script("arguments[0].click();", more_btn)
                time.sleep(1)
            else:
                break
        except:
            break

def scroll_to_bottom():
    """스크롤 내려서 전체 메뉴 로딩"""
    last_height = driver.execute_script("return document.body.scrollHeight")
    for _ in range(3):
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(1)
        new_height = driver.execute_script("return document.body.scrollHeight")
        if new_height == last_height:
            break
        last_height = new_height

for brand, keyword in search_queries:
    try:
        print(f"\n[+] {brand} 메뉴 확인 시작")
        driver.get(f"https://map.naver.com/p/search/{keyword}")
        time.sleep(5)

        driver.switch_to.default_content()
        iframe = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "iframe#entryIframe")))
        driver.switch_to.frame(iframe)

        # 메뉴 탭 클릭
        tab_clicked = False
        tabs = driver.find_elements(By.CSS_SELECTOR, "a._tab-menu, nav a, ul li a")
        for tab in tabs:
            if "메뉴" in tab.text:
                driver.execute_script("arguments[0].click();", tab)
                time.sleep(3)
                tab_clicked = True
                break
        if not tab_clicked:
            print("❌ 메뉴 탭 없음")
            continue

        time.sleep(2)
        click_more_buttons()
        scroll_to_bottom()
        time.sleep(2)

        # 메뉴 영역 찾기
        all_sections = driver.find_elements(By.CSS_SELECTOR, "div.naver_order_contents div.order_list_inner")
        print(f"음식 구분 {len(all_sections)}개")

        total_count = 0
        for section in all_sections:
            try:
                # 음식구분 이름 (예: 신메뉴, 치킨버거 등)
                section_title = section.find_element(By.CSS_SELECTOR, "div.order_category_title span").text.strip()
            except:
                section_title = "카테고리 없음"

            items = section.find_elements(By.CSS_SELECTOR, "ul.order_list_area > li")
            section_menus = []

            for item in items:
                try:
                    name = item.find_element(By.CSS_SELECTOR, "div.MenuContent__tit__313LA").text.strip()
                    if not ("버거" in name or "세트" in name):
                        continue

                    try:
                        desc = item.find_element(By.CSS_SELECTOR, "div.MenuContent__detail__OG635 span.detail_txt").text.strip()
                    except:
                        desc = "설명 없음"

                    try:
                        price = item.find_element(By.CSS_SELECTOR, "div.MenuContent__price__lhCy9 strong").text.strip()
                    except:
                        price = "가격 없음"

                    section_menus.append((name, price, desc))

                except:
                    continue

            if section_menus:
                print(f"- {section_title} ({len(section_menus)}개)")
                for m in section_menus:
                    print(f"  · {m[0]} / {m[1]} / {m[2]}")
                total_count += len(section_menus)

        print(f"총 메뉴 수: {total_count}")

    except Exception as e:
        print(f"[!] {brand} 오류: {e}")

# driver.quit()
```

```python
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import TimeoutException, NoSuchElementException

search_queries = [
    ("쉐이크쉑", "쉐이크쉑 신림점"),
    ("맥도날드", "맥도날드 신림점"),
    ("롯데리아", "롯데리아 신림역점"),
    ("버거킹", "버거킹 신림역점"),
    ("버거운버거", "버거운버거 서울대점"),
    ("KFC", "KFC 신림역"),
    ("노브랜드버거", "노브랜드버거 신림남부점"),
    ("프랭크버거", "프랭크버거 신림녹두거리점"),
    ("움버거앤윙스", "움버거앤윙스 신림역점"),
    ("버거리", "버거리 낙성대점")
]

# Chrome 옵션 설정 (포트 고정 및 안정성 향상)
options = webdriver.ChromeOptions()
options.add_argument("--start-maximized")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("--disable-gpu")
options.add_argument("--remote-debugging-port=9222")  # 포트 고정
options.add_argument("--disable-web-security")
options.add_argument("--allow-running-insecure-content")
options.add_argument("--disable-extensions")
options.add_argument("--disable-plugins")
options.add_argument("--disable-images")  # 이미지 로딩 비활성화로 속도 향상
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)

# User-Agent 설정 (봇 감지 방지)
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36")

try:
    driver = webdriver.Chrome(options=options)
    wait = WebDriverWait(driver, 15)  # 대기 시간 증가
    
    # 봇 감지 방지 스크립트
    driver.execute_script("Object.defineProperty(navigator, 'webdriver', {get: () => undefined})")
    
    print("✅ Chrome 브라우저가 성공적으로 시작되었습니다.")
    print(f"🔧 디버깅 포트: 9222")
    
except Exception as e:
    print(f"❌ Chrome 브라우저 시작 실패: {e}")
    print("💡 해결방법:")
    print("   1. 기존 Chrome 프로세스 종료")
    print("   2. ChromeDriver 버전 확인")
    print("   3. 관리자 권한으로 실행")
    exit()

total_global_count = 0
total_global_menu_details = []

def click_all_more_buttons():
    """모든 더보기 버튼을 클릭"""
    more_buttons_clicked = 0
    try:
        more_buttons = driver.find_elements(By.CSS_SELECTOR, ".fvwqf[role='button']")
        for btn in more_buttons:
            try:
                if btn.is_displayed():
                    driver.execute_script("arguments[0].click();", btn)
                    time.sleep(1)
                    more_buttons_clicked += 1
            except:
                continue
    except:
        pass
    return more_buttons_clicked

def scroll_and_load_all_content():
    """스크롤하고 더보기 버튼을 클릭해서 모든 컨텐츠 로드"""
    max_attempts = 5
    for attempt in range(max_attempts):
        # 스크롤 다운
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(2)
        
        # 더보기 버튼 클릭
        more_clicked = click_all_more_buttons()
        
        # 추가 스크롤
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(2)
        
        if more_clicked == 0:
            break
    
    # 최종 스크롤
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(2)

for brand, keyword in search_queries:
    try:
        print(f"\n{'='*60}")
        print(f"🍔 {brand} 메뉴 분석 시작")
        print(f"🔍 검색어: {keyword}")
        print(f"{'='*60}")
        
        driver.get(f"https://map.naver.com/p/search/{keyword}")
        time.sleep(5)

        # 검색 결과 확인
        try:
            # 검색 결과가 없는 경우 체크
            no_result = driver.find_elements(By.CSS_SELECTOR, ".place_didntmatch, .no_result")
            if no_result:
                print("❌ 검색 결과가 없습니다.")
                continue
        except:
            pass

        # iframe 전환
        driver.switch_to.default_content()
        try:
            iframe = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "iframe#entryIframe")))
            driver.switch_to.frame(iframe)
        except TimeoutException:
            print("❌ 매장 상세 페이지로 진입할 수 없습니다.")
            continue

        # 매장 이름 확인
        try:
            store_name = driver.find_element(By.CSS_SELECTOR, "h1.GHAhO, .place_name, span.Fc1rA").text
            print(f"🏪 매장명: {store_name}")
        except:
            print("⚠️ 매장명을 확인할 수 없습니다.")

        # 메뉴 탭 찾기 및 클릭
        tab_clicked = False
        print("🔍 메뉴 탭 찾는 중...")
        
        # 다양한 메뉴 탭 선택자 시도
        menu_selectors = [
            "a._tab-menu:contains('메뉴')",
            "nav a:contains('메뉴')", 
            "ul li a:contains('메뉴')",
            "a[data-tab='menu']",
            ".flicking-camera a:contains('메뉴')",
            "._tab-menu[data-code='MENU']"
        ]
        
        for selector in menu_selectors:
            try:
                if ":contains(" in selector:
                    # contains는 CSS에서 지원하지 않으므로 JavaScript로 찾기
                    tabs = driver.find_elements(By.CSS_SELECTOR, "a._tab-menu, nav a, ul li a")
                    for tab in tabs:
                        if "메뉴" in tab.text:
                            driver.execute_script("arguments[0].click();", tab)
                            time.sleep(3)
                            tab_clicked = True
                            print("✅ 메뉴 탭을 찾았습니다.")
                            break
                else:
                    tab = driver.find_element(By.CSS_SELECTOR, selector)
                    driver.execute_script("arguments[0].click();", tab)
                    time.sleep(3)
                    tab_clicked = True
                    print("✅ 메뉴 탭을 찾았습니다.")
                    break
            except:
                continue
            
            if tab_clicked:
                break
        
        if not tab_clicked:
            print("❌ 메뉴 탭을 찾을 수 없습니다.")
            print("💡 이 매장은 메뉴 정보가 등록되지 않았을 수 있습니다.")
            continue

        # 메뉴 컨텐츠 로딩 대기
        time.sleep(3)
        
        # 메뉴 컨테이너 확인
        try:
            menu_check = driver.find_elements(By.CSS_SELECTOR, "div.naver_order_contents, .menu_list, .order_list")
            if not menu_check:
                print("❌ 메뉴 컨텐츠를 찾을 수 없습니다.")
                continue
        except:
            print("❌ 메뉴 페이지 로딩에 실패했습니다.")
            continue

        # 모든 컨텐츠 로드
        scroll_and_load_all_content()

        # 전체 메뉴 구성 컨테이너 찾기
        menu_containers = driver.find_elements(By.CSS_SELECTOR, "div.naver_order_contents")
        print(f"📦 전체 메뉴 컨테이너 수: {len(menu_containers)}개")

        total_brand_count = 0
        brand_menu_details = []

        # 각 메뉴 컨테이너 처리
        for container_idx, container in enumerate(menu_containers):
            print(f"\n📋 메뉴 컨테이너 {container_idx + 1}:")
            
            # 음식 구분 찾기
            food_sections = container.find_elements(By.CSS_SELECTOR, "div.order_list_inner")
            print(f"   🍽 음식 구분 수: {len(food_sections)}개")

            # 각 음식 구분 처리
            for section_idx, section in enumerate(food_sections):
                try:
                    # 음식 구분 제목 가져오기
                    section_title = "카테고리 정보 없음"
                    try:
                        section_title = section.find_element(By.CSS_SELECTOR, "div.order_category_title span").text.strip()
                    except:
                        section_title = f"카테고리 {section_idx + 1}"

                    # ul 요소들 찾기
                    ul_elements = section.find_elements(By.CSS_SELECTOR, "ul.order_list_area")
                    print(f"      └ '{section_title}' - UL 수: {len(ul_elements)}개")

                    section_menu_count = 0
                    
                    # 각 ul 내의 li 요소들 처리
                    for ul_idx, ul in enumerate(ul_elements):
                        li_elements = ul.find_elements(By.CSS_SELECTOR, "li")
                        print(f"         └ UL {ul_idx + 1}: {len(li_elements)}개 메뉴")
                        
                        # 각 메뉴 아이템 처리
                        for li in li_elements:
                            try:
                                # 메뉴 정보 컨테이너 찾기
                                menu_info = li.find_element(By.CSS_SELECTOR, "div.MenuContent__info_detail__rCviz")
                                
                                # 메뉴명
                                menu_name = "메뉴명 없음"
                                try:
                                    menu_name = menu_info.find_element(By.CSS_SELECTOR, "div.MenuContent__tit__313LA").text.strip()
                                except:
                                    continue  # 메뉴명이 없으면 스킵
                                
                                # 메뉴 설명
                                menu_desc = ""
                                try:
                                    desc_elem = menu_info.find_element(By.CSS_SELECTOR, "div.MenuContent__detail__OG635 span.detail_txt")
                                    menu_desc = desc_elem.text.strip()
                                except:
                                    pass
                                
                                # 가격
                                menu_price = "가격 없음"
                                try:
                                    price_elem = menu_info.find_element(By.CSS_SELECTOR, "div.MenuContent__price__lhCy9 strong")
                                    menu_price = price_elem.text.strip()
                                except:
                                    pass
                                
                                # 버거/세트 필터링
                                text_to_check = f"{menu_name} {menu_desc}".lower()
                                if "버거" in text_to_check or "세트" in text_to_check:
                                    # 메뉴 정보 저장
                                    menu_detail = {
                                        'brand': brand,
                                        'category': section_title,
                                        'name': menu_name,
                                        'price': menu_price,
                                        'description': menu_desc
                                    }
                                    
                                    brand_menu_details.append(menu_detail)
                                    section_menu_count += 1
                                
                            except Exception as menu_err:
                                continue
                    
                    if section_menu_count > 0:
                        print(f"         → {section_title}: {section_menu_count}개 버거/세트 메뉴")
                    
                    total_brand_count += section_menu_count
                    
                except Exception as section_err:
                    print(f"      ❌ 섹션 처리 오류: {section_err}")
                    continue

        # 브랜드별 결과 출력
        print(f"\n🎯 {brand} 총 버거/세트 메뉴 수: {total_brand_count}개")
        print(f"📋 {brand} 버거/세트 메뉴 목록:")
        
        for menu in brand_menu_details:
            print(f"   • {menu['name']} ({menu['category']}) - {menu['price']}")
            if menu['description']:
                print(f"     └ {menu['description']}")
        
        total_global_count += total_brand_count
        total_global_menu_details.extend(brand_menu_details)

    except Exception as e:
        print(f"❌ {brand} 전체 처리 오류: {e}")
        print(f"💡 원인: 매장이 존재하지 않거나, 메뉴 정보가 없거나, 페이지 구조가 다를 수 있습니다.")

# 최종 결과 출력
print(f"\n{'='*60}")
print(f"🌟 전체 브랜드 총 버거/세트 메뉴 수: {total_global_count}개")
print(f"📊 성공한 브랜드: {len([menu['brand'] for menu in total_global_menu_details if menu])}개")
print(f"{'='*60}")

# 브랜드별 요약
brand_summary = {}
for menu in total_global_menu_details:
    brand = menu['brand']
    if brand not in brand_summary:
        brand_summary[brand] = 0
    brand_summary[brand] += 1

print("\n📊 브랜드별 버거/세트 메뉴 수 요약:")
for brand, count in brand_summary.items():
    print(f"   {brand}: {count}개")

# 안전한 브라우저 종료
# try:
#     driver.quit()
#     print("\n🔒 브라우저가 안전하게 종료되었습니다.")
# except:
#     print("\n⚠️ 브라우저 종료 중 오류 발생")

# Chrome 프로세스 강제 종료 (Windows 기준)
import os
import platform

if platform.system() == "Windows":
    try:
        os.system("taskkill /f /im chrome.exe /t")
        os.system("taskkill /f /im chromedriver.exe /t")
        print("🧹 Chrome 프로세스 정리 완료")
    except:
        pass
```
