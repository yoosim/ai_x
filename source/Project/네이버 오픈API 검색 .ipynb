```python
# 저장된 텍스트 파일 불러오기 
with open('0709_데이터 폴더/01.메뉴명 추출/brand_menu_list(브랜드명+메뉴명).txt', 'r', encoding='utf-8')as f:
    menu_list = [line.strip() for line in f.readlines()]


# print(len(menu_list))

print(f"원본 메뉴 수 : {len(menu_list)}")

# 특정 단어 포함된 메뉴 전체 삭제
print("========1. 특정 단어 포함 메뉴 전체 삭제 ============")

# 삭제할 단어들
delete_words = ['대표', '캔', '펩시','탄산', '오렌지','사이다','콜라','메뉴','리뷰','사진','정보','아이스티','캐모마일','에이드','아메리카노','아이스초코','핫초코','닥터페퍼'] # 대표라고 딸려온 텍스트 메뉴 삭제 
filtered_menu = []
deleted_menu = []

for menu in menu_list:
    # 삭제할 단어가 포함되어 있는지 확인
    checked_delete = any(word in menu for word in delete_words)
    

    if checked_delete:
        deleted_menu.append(menu)
#         print(f"삭제됨 :{menu}")
    else:
        filtered_menu.append(menu)
# display(menu_list)        
print(f"삭제된 메뉴 수 : {len(deleted_menu)}")
print(f"남은 메뉴 수 : {len(filtered_menu)}")
display(filtered_menu)
```
