```python
# 필요없는 이미지 데이터 폴더 삭제 
import os
import shutil

root_path = '0709_데이터 폴더-카피본'

# 삭제 기준 키워드 
exclude_keywords=[
    '대표', '캔', '펩시','탄산', 
    '오렌지','사이다','콜라','메뉴',
    '리뷰','사진','정보','아이스티',
    '캐모마일','에이드','아메리카노',
    '아이스초코','핫초코','닥터페퍼',
    '캔','인기메뉴','대표메뉴',
    '레몬에이드','인기신규','콘샐러드',
    '콘','너겟','아이스크림','링','프렌치프라이',
    '조각','윙','지파이','강정','케이준','반마리',
    '한마리','허브순살치킨런(M)','쥐포','순살후라이드']

# 삭제 로그 남길 리스트
deleted_folders = []

# 브랜드 폴더 체크
for brand in os.listdir(root_path):
    brand_path = os.path.join(root_path, brand)
    if not os.path.isdir(brand_path):
        continue
        
    #  메뉴 폴더 순회
    for menu in os.listdir(brand_path):
        menu_path = os.path.join(brand_path,menu)
        if not os.path.isdir(menu_path):
            continue
            
        #메뉴 폴더 키워드 포함시 삭제
        if any(keyword in menu.lower() for keyword in exclude_keywords):
            print(f"삭제중 : {menu_path}")
            shutil.rmtree(menu_path) #  폴더 전체 삭제
            deleted_folders.append(menu_path)
            
# 결과 출력
print('\n 삭제된 폴더 목록 :')
for path in deleted_folders:
    print(path)
```

```python
# 폴더명 ()괄호안 영문 지우기 / []대괄호 지우기
import os
import re

# 루트 폴더 
root_path = '0709_데이터 폴더-카피본'

#  로그 기록
renamed_folders=[]

#정규표현식 패턴
pattern_parenthess_eng = re.compile(r'\((?=[A-Za-z\s]*\))[^)]*\)')
pattern_brackets_all = re.compile(r'\[.*?\]')  # 대괄호 전체

# 브랜드 폴더 체크 
for brand in os.listdir(root_path):
    brand_path = os.path.join(root_path, brand)
    if not os.path.isdir(brand_path):
        continue 
        
    for menu in os.listdir(brand_path):
        menu_path = os.path.join(brand_path, menu)
        if not os.path.isdir(menu_path):
            continue
            
        # 1.괄호()안에 영어만 삭제 
        new_menu = pattern_parenthess_eng.sub('',menu)
        
        # 2.대괄호[] 전체 삭제 
        new_menu = pattern_brackets_all.sub('',menu)
        
        # 3. 공백 정리 
        new_menu = re.sub(r'\s+',' ', new_menu).strip()
        
        if new_menu != menu :
            new_menu_path = os.path.join(brand_path,new_menu)
            
            # 중복된 폴더가 없다면 이름 변경
            if not os.path.exists(new_menu_path):
                os.rename(menu_path, new_menu_path)
                renamed_folders.append((menu,new_menu))
            else:
                print(f"이미 존재 하는 폴더 : {new_menu_path}, 스킵")
                
print('\n 변경된 폴더명 :')
for old,new in renamed_folders:
    print(f"{old} -> {new}")
```

```python
# 폴더명 + 메뉴이름으로 text 파일 만들기 (json -> 딕셔너리-> text)
import json

# 브랜드명_ 메뉴 폴더 경로
root_path = '0709_데이터 폴더-카피본'

# 최종 저장 경로
output_dir = 'search_text(0710_15)'
os.makedirs(output_dir,exist_ok=True)

# 파일 변환용 담기
brand_menu_dict ={}
flat_search_terms = []

# 브랜드 폴더 체크
for brand in os.listdir(root_path):
    brand_path = os.path.join(root_path,brand)
    if not os.path.isdir(brand_path):
        continue
        
    menu_list = []
    for menu in os.listdir(brand_path):
        menu_path = os.path.join(brand_path, menu)
        if not os.path.isdir(menu_path):
            continue
        menu_list.append(menu)
        flat_search_terms.append(f"{brand} {menu}")
    if menu_list:
        brand_menu_dict[brand] = menu_list
        
# json으로 저장
with open(os.path.join(output_dir,'brand_menu.json'), 'w', encoding='utf-8')as f:
    json.dump(brand_menu_dict,f, ensure_ascii=False, indent=2)
    
# dict으로 저장
with open(os.path.join(output_dir,'brand_menu_dict.py'),'w',encoding='utf-8')as f:
    f.write('brand_menu_dict =')
    json.dump(brand_menu_dict, f, ensure_ascii=False, indent=2)

# text 파일로 저장
joined_line = ','.join(flat_search_terms)
with open(os.path.join(output_dir, 'search_keyword.txt'),'w',encoding='utf-8') as f:
    f.write(joined_line)
    
print("✅ 저장 완료!")
print(f"- brand_menu.json")
print(f"- brand_menu_dict.py")
print(f"- search_keywords.txt (쉼표 연결 형식)")
```
