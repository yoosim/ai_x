```
1. csv 데이터 불러오기 
2. 임베딩
3. 사용자 입력받기 
4. 유사도 계산
5. csv 파일로 저장

```

````python
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import pickle
import random

class DatingChatbot:
    def __init__(self, csv_path):
        """
        소개팅 연습 챗봇 초기화
        """
        # 1. 모델 로드 (한국어 특화)
        print("모델 로딩 중...")
        self.model = SentenceTransformer('jhgan/ko-sroberta-multitask')
        
        # 2. CSV 데이터 로드
        print("데이터 로딩 중...")
        self.df = pd.read_csv('data/dating_circular_talkset_10k_cp949 (1).csv', encoding='cp949')
        
        # 3. 대화 상태 관리
        self.target_profile = {}  # 소개팅 대상 정보
        self.conversation_history = []  # 대화 기록
        self.learned_profile = {}  # 대화 중 학습한 정보
        
        # 4. 임베딩 생성 또는 로드
        self.embeddings = None
        self.load_or_create_embeddings()
    
    def load_or_create_embeddings(self):
        """
        임베딩 파일이 있으면 로드, 없으면 생성
        """
        try:
            # 기존 임베딩 파일 로드 시도
            with open('question_embeddings.pkl', 'rb') as f:
                self.embeddings = pickle.load(f)
            print("기존 임베딩 파일을 로드했습니다.")
        except FileNotFoundError:
            # 임베딩 새로 생성
            print("임베딩 생성 중... (시간이 오래 걸릴 수 있습니다)")
            questions = self.df['질문'].tolist()
            self.embeddings = self.model.encode(questions, show_progress_bar=True)
            
            # 임베딩 저장
            with open('question_embeddings.pkl', 'wb') as f:
                pickle.dump(self.embeddings, f)
            print("임베딩 생성 완료 및 저장!")
    
    def set_target_profile(self, profile_dict):
        """
        소개팅 대상 프로필 설정
        예: {"취미": "영화감상", "성격": "외향적", "직업": "개발자"}
        """
        self.target_profile = profile_dict
        print(f"타겟 프로필 설정: {profile_dict}")
    
    def find_similar_questions(self, user_question, top_k=3):
        """
        사용자 질문과 유사한 질문들 찾기
        """
        # 사용자 질문 임베딩
        user_embedding = self.model.encode([user_question])
        
        # 코사인 유사도 계산
        similarities = cosine_similarity(user_embedding, self.embeddings)[0]
        
        # 상위 k개 인덱스
        top_indices = np.argsort(similarities)[::-1][:top_k]
        
        # 결과 반환
        results = []
        for idx in top_indices:
            results.append({
                'index': idx,
                'similarity': similarities[idx],
                'question': self.df.iloc[idx]['질문'],
                'answer': self.df.iloc[idx]['답변'],
                'follow_up': self.df.iloc[idx]['리액션_추가질문']
            })
        
        return results
    
    def customize_answer(self, base_answer, question_topic):
        """
        타겟 프로필에 맞게 답변 커스터마이징
        """
        # 프로필 정보가 있으면 활용
        complete_profile = {**self.target_profile, **self.learned_profile}
        
        # 간단한 키워드 매칭으로 답변 개인화
        customized = base_answer
        
        # 예시: 영화 관련 질문일 때
        if any(word in question_topic for word in ['영화', '드라마', '볼거리']):
            if '영화' in complete_profile:
                customized = f"{base_answer} 특히 {complete_profile['영화']} 장르를 좋아해요."
        
        # 취미 관련
        if '취미' in question_topic:
            if '취미' in complete_profile:
                customized = f"{complete_profile['취미']}를 좋아해요. {base_answer}"
        
        return customized
    
    def extract_info_from_conversation(self, user_input):
        """
        사용자 답변에서 정보 추출하여 학습된 프로필에 추가
        (간단한 키워드 기반 - 실제로는 더 정교한 NLP 필요)
        """
        # 간단한 패턴 매칭
        if '좋아' in user_input:
            if any(word in user_input for word in ['영화', '드라마']):
                self.learned_profile['선호_엔터테인먼트'] = '영화/드라마'
            elif any(word in user_input for word in ['운동', '헬스', '축구']):
                self.learned_profile['선호_활동'] = '운동'
    
    def chat(self, user_input):
        """
        메인 대화 함수
        """
        # 1. 사용자 입력에서 정보 학습
        self.extract_info_from_conversation(user_input)
        
        # 2. 유사한 질문 찾기
        similar_questions = self.find_similar_questions(user_input, top_k=1)
        best_match = similar_questions[0]
        
        # 3. 답변 커스터마이징
        customized_answer = self.customize_answer(
            best_match['answer'], 
            best_match['question']
        )
        
        # 4. 대화 기록에 추가
        self.conversation_history.append({
            'user': user_input,
            'bot': customized_answer,
            'follow_up': best_match['follow_up'],
            'similarity': best_match['similarity']
        })
        
        # 5. 답변과 추가 질문 반환
        response = {
            'answer': customized_answer,
            'follow_up_question': best_match['follow_up'],
            'similarity_score': best_match['similarity'],
            'learned_profile': self.learned_profile
        }
        
        return response
    
    def get_conversation_summary(self):
        """
        지금까지의 대화 요약
        """
        return {
            'total_exchanges': len(self.conversation_history),
            'target_profile': self.target_profile,
            'learned_profile': self.learned_profile,
            'conversation_history': self.conversation_history[-3:]  # 최근 3개만
        }
    
    def save_conversation_to_csv(self, filename=None):
        """
        대화 결과를 CSV 파일로 저장 (한글)
        """
        from datetime import datetime
        
        if filename is None:
            timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
            filename = f"소개팅연습결과_{timestamp}.csv"
        
        # 대화 데이터 정리
        conversation_data = []
        
        # 첫 번째 행: 기본 정보
        conversation_data.append({
            '순번': 0,
            '항목': '연습일시',
            '내용': datetime.now().strftime("%Y년 %m월 %d일 %H시 %M분"),
            '상세정보': '',
            '유사도점수': '',
            '비고': '소개팅 연습 시작'
        })
        
        # 타겟 프로필 정보
        for idx, (key, value) in enumerate(self.target_profile.items(), 1):
            conversation_data.append({
                '순번': idx,
                '항목': '설정된 상대방 정보',
                '내용': f"{key}: {value}",
                '상세정보': '',
                '유사도점수': '',
                '비고': '초기 설정'
            })
        
        # 대화 내용
        for idx, conv in enumerate(self.conversation_history, len(self.target_profile) + 2):
            # 사용자 질문
            conversation_data.append({
                '순번': idx * 2 - 1,
                '항목': '내가 한 질문',
                '내용': conv['user'],
                '상세정보': '',
                '유사도점수': f"{conv['similarity']:.3f}",
                '비고': f"{idx-len(self.target_profile)-1}번째 대화"
            })
            
            # 챗봇 답변
            conversation_data.append({
                '순번': idx * 2,
                '항목': '상대방 답변',
                '내용': conv['bot'],
                '상세정보': conv['follow_up'],
                '유사도점수': '',
                '비고': '챗봇 응답'
            })
        
        # 학습된 정보
        if self.learned_profile:
            for idx, (key, value) in enumerate(self.learned_profile.items()):
                conversation_data.append({
                    '순번': len(conversation_data) + 1,
                    '항목': '대화중 파악된 정보',
                    '내용': f"{key}: {value}",
                    '상세정보': '',
                    '유사도점수': '',
                    '비고': '자동 학습'
                })
        
        # DataFrame으로 변환하여 저장
        df_result = pd.DataFrame(conversation_data)
        df_result.to_csv(filename, index=False, encoding='utf-8-sig')  # 한글 깨짐 방지
        
        print(f"\n✅ 대화 결과가 '{filename}' 파일로 저장되었습니다!")
        return filename

# 사용 예시
def main():
    # 1. 챗봇 초기화
    chatbot = DatingChatbot('your_data.csv')  # CSV 파일 경로
    
    # 2. 타겟 프로필 설정
    chatbot.set_target_profile({
        "취미": "영화감상", 
        "성격": "외향적", 
        "직업": "마케터"
    })
    
    # 3. 대화 시작
    print("=== 소개팅 연습 챗봇 ===")
    print("종료하려면 'quit'를 입력하세요.\n")
    
    while True:
        user_input = input("당신: ")
        
        if user_input.lower() == 'quit':
            break
        
        # 챗봇 응답
        response = chatbot.chat(user_input)
        
        print(f"상대방: {response['answer']}")
        print(f"추가질문: {response['follow_up_question']}")
        print(f"유사도: {response['similarity_score']:.3f}")
        print(f"학습된 정보: {response['learned_profile']}")
        print("-" * 50)
    
    # 4. 대화 요약 및 저장
    summary = chatbot.get_conversation_summary()
    print("\n=== 대화 요약 ===")
    print(f"총 대화 횟수: {summary['total_exchanges']}")
    print(f"학습된 정보: {summary['learned_profile']}")
    
    # 5. CSV 파일로 저장
    save_choice = input("\n대화 결과를 CSV 파일로 저장하시겠습니까? (y/n): ")
    if save_choice.lower() in ['y', 'yes', '예', 'ㅇ']:
        filename = chatbot.save_conversation_to_csv()
        print(f"파일이 저장되었습니다: {filename}")
    
    print("\n소개팅 연습을 마칩니다. 수고하셨습니다! 💕")

if __name__ == "__main__":
    main()
````
